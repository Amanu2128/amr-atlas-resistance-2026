```python
# ============================================================
# CELL 12 (REVISED): SEPARATE FEATURE SETS PER TARGET
# ============================================================
# exclude ITS OWN target drug family, and may legitimately use
# the OTHER drug family as a co-resistance signal.
# ============================================================

# --- The 7 original epidemiological/surveillance features ---
surveillance_features = ["Species", "Continent", "Source_Category", "Speciality", "Age Group", "Gender", "Year"]

# --- The 16 co-resistance drugs from unrelated drug classes (unchanged) ---
coresistance_features = [
    "Amikacin_I", "Amoxycillin clavulanate_I", "Ampicillin_I", "Ampicillin sulbactam_I",
    "Azithromycin_I", "Aztreonam_I", "Cefepime_I", "Cefiderocol_I", "Ciprofloxacin_I",
    "Colistin_I", "Gentamicin_I", "Levofloxacin_I", "Minocycline_I",
    "Piperacillin tazobactam_I", "Tigecycline_I", "Trimethoprim sulfa_I"
]

# --- The two target-defining drug families (each only excluded from its OWN model) ---
carbapenem_drug_cols = ["Meropenem_I", "Imipenem_I", "Doripenem_I", "Ertapenem_I"]
cephalosporin_drug_cols = ["Ceftriaxone_I", "Ceftazidime_I", "Cefpodoxime_I", "Cefixime_I", "Ceftibuten_I"]

# ------------------------------------------------------------
# THREE FEATURE SETS, PER THE REVIEWER'S RECOMMENDATION
# ------------------------------------------------------------

# 1. Surveillance-only (no antibiotics at all) — same for both targets
approved_features_surveillance = surveillance_features.copy()

# 2. Carbapenem co-resistance model — excludes ONLY carbapenems,
#    legitimately includes cephalosporins as co-resistance signal
approved_features_carbapenem = surveillance_features + coresistance_features + cephalosporin_drug_cols

# 3. Cephalosporin co-resistance model — excludes ONLY cephalosporins,
#    legitimately includes carbapenems as co-resistance signal
approved_features_cephalosporin = surveillance_features + coresistance_features + carbapenem_drug_cols

# ------------------------------------------------------------
# SAFETY CHECKS
# ------------------------------------------------------------
overlap_carb = set(carbapenem_drug_cols) & set(approved_features_carbapenem)
overlap_ceph = set(cephalosporin_drug_cols) & set(approved_features_cephalosporin)

print("=" * 70)
print("THREE FEATURE SETS DEFINED")
print("=" * 70)
print(f"\n1. Surveillance-only:        {len(approved_features_surveillance)} features")
print(f"2. Carbapenem co-resistance: {len(approved_features_carbapenem)} features")
print(f"3. Cephalosporin co-resist.: {len(approved_features_cephalosporin)} features")

print(f"\nLeakage check (carbapenem model): "
      f"{'❌ FAILED' if overlap_carb else '✅ PASSED — carbapenem drugs correctly excluded'}")
print(f"Leakage check (cephalosporin model): "
      f"{'❌ FAILED' if overlap_ceph else '✅ PASSED — cephalosporin drugs correctly excluded'}")

print("\nConfirming co-resistance is asymmetric (correct behavior):")
print(f"  Cephalosporin drugs used as features in CARBAPENEM model: "
      f"{set(cephalosporin_drug_cols).issubset(set(approved_features_carbapenem))}")
print(f"  Carbapenem drugs used as features in CEPHALOSPORIN model: "
      f"{set(carbapenem_drug_cols).issubset(set(approved_features_cephalosporin))}")
```

```python
# ============================================================
# CELL 14 : DATA-DRIVEN SPARSE-COLUMN REMOVAL
# + "NOT TESTED" FILL + FREQUENCY SUMMARY
# ============================================================
# Note: targets (Carbapenem_Resistant, Cephalosporin_Resistant)
# were already constructed in Cell 10. Everything below affects
# ONLY predictor/feature columns, never the targets themselves.

# --- Step 1: remove the 100%-missing column (already agreed) ---
coresistance_features = [c for c in coresistance_features if c != "Azithromycin_I"]

# --- Step 2: compute missingness for every antibiotic column,
#     apply ONE clearly stated threshold consistently ---
all_antibiotic_cols = list(set(
    coresistance_features + carbapenem_drug_cols + cephalosporin_drug_cols
))

MISSING_THRESHOLD = 90.0   # explicit, single, defensible cutoff

missing_pct = (df_gn[all_antibiotic_cols].isna().mean() * 100).round(2)
sparse_cols_to_drop = missing_pct[missing_pct >= MISSING_THRESHOLD].index.tolist()

print("=" * 70)
print(f"AUTOMATIC SPARSE-COLUMN REMOVAL (threshold: ≥{MISSING_THRESHOLD}% missing)")
print("=" * 70)
print("\nColumns being dropped:")
for c in sparse_cols_to_drop:
    print(f"  - {c}: {missing_pct[c]}% missing")

# --- Step 3: remove dropped columns from every relevant list ---
coresistance_features = [c for c in coresistance_features if c not in sparse_cols_to_drop]
carbapenem_drug_cols_clean = [c for c in carbapenem_drug_cols if c not in sparse_cols_to_drop]
cephalosporin_drug_cols_clean = [c for c in cephalosporin_drug_cols if c not in sparse_cols_to_drop]

approved_features_surveillance = surveillance_features.copy()
approved_features_carbapenem = surveillance_features + coresistance_features + cephalosporin_drug_cols_clean
approved_features_cephalosporin = surveillance_features + coresistance_features + carbapenem_drug_cols_clean

print(f"\nCarbapenem model features:    {len(approved_features_carbapenem)}")
print(f"Cephalosporin model features: {len(approved_features_cephalosporin)}")

# --- Step 4: fill remaining antibiotic columns with "Not Tested" ---
remaining_antibiotic_cols = list(set(
    coresistance_features + carbapenem_drug_cols_clean + cephalosporin_drug_cols_clean
))

for col in remaining_antibiotic_cols:
    df_gn[col] = df_gn[col].fillna("Not Tested")

print(f"\n'Not Tested' applied to {len(remaining_antibiotic_cols)} remaining antibiotic columns.")

# --- Step 5: frequency summary — S/I/R/Not Tested counts per drug ---
summary_rows = []
for col in remaining_antibiotic_cols:
    counts = df_gn[col].value_counts()
    summary_rows.append({
        "Drug": col,
        "Susceptible": counts.get("Susceptible", 0),
        "Intermediate": counts.get("Intermediate", 0),
        "Resistant": counts.get("Resistant", 0),
        "Not Tested": counts.get("Not Tested", 0),
    })

summary_df = pd.DataFrame(summary_rows).sort_values("Not Tested", ascending=False)

print("\n" + "=" * 70)
print("FREQUENCY SUMMARY — REMAINING ANTIBIOTIC COLUMNS")
print("=" * 70)
display(summary_df)
```

```python
# ============================================================
# CELL 15: BUILD FINAL MODELING TABLES (ONE PER TARGET)
# ============================================================

# --- Model A: Carbapenem resistance ---
model_data_carb = df_gn[["Carbapenem_Resistant"] + approved_features_carbapenem].copy()
before_carb = len(model_data_carb)
model_data_carb = model_data_carb.dropna(subset=["Carbapenem_Resistant"])
after_carb = len(model_data_carb)

# --- Model B: Cephalosporin resistance ---
model_data_ceph = df_gn[["Cephalosporin_Resistant"] + approved_features_cephalosporin].copy()
before_ceph = len(model_data_ceph)
model_data_ceph = model_data_ceph.dropna(subset=["Cephalosporin_Resistant"])
after_ceph = len(model_data_ceph)

print("=" * 70)
print("FINAL MODELING TABLES BUILT")
print("=" * 70)
print(f"\nModel A (Carbapenem):")
print(f"  Rows before dropping missing target: {before_carb:,}")
print(f"  Rows after:                          {after_carb:,}")
print(f"  Columns (features + target):         {model_data_carb.shape[1]}")

print(f"\nModel B (Cephalosporin):")
print(f"  Rows before dropping missing target: {before_ceph:,}")
print(f"  Rows after:                          {after_ceph:,}")
print(f"  Columns (features + target):         {model_data_ceph.shape[1]}")

print("\n--- Confirm zero remaining missing values anywhere ---")
print("Model A missing values total:", model_data_carb.isna().sum().sum())
print("Model B missing values total:", model_data_ceph.isna().sum().sum())
```

```python
# ============================================================
# CELL 17: ENCODING — FIT ON TRAINING DATA ONLY
# ============================================================
# Three encoding strategies, chosen per variable type:
#  1. Age Group        -> ordinal (true order respected)
#  2. Antibiotic S/I/R  -> ordinal (clinical severity order respected)
#  3. Everything else   -> label encoding (no natural order exists)
#
# All encoders are fit ONLY on training years (2004-2019).
# Unseen categories in the test set become "Unknown" (encoded
# as -1) rather than being dropped -- per agreed correction.

from sklearn.preprocessing import LabelEncoder

# ------------------------------------------------------------
# Define the ordinal orders explicitly
# ------------------------------------------------------------
age_order = ["0 - 17", "18 - 30", "31 - 60", "61+", "Unknown"]
sir_order = ["Susceptible", "Intermediate", "Resistant", "Not Tested"]

def encode_ordinal(train_df, test_df, col, order):
    order_map = {cat: i for i, cat in enumerate(order)}
    train_df[col + "_encoded"] = train_df[col].map(order_map)
    test_df[col + "_encoded"] = test_df[col].map(order_map)
    # Anything genuinely unmapped (shouldn't happen, but safe fallback)
    test_df[col + "_encoded"] = test_df[col + "_encoded"].fillna(-1)
    return order_map

def encode_nominal_with_unknown(train_df, test_df, col):
    le = LabelEncoder()
    le.fit(train_df[col])
    train_df[col + "_encoded"] = le.transform(train_df[col])
    known = set(le.classes_)
    test_df[col + "_encoded"] = test_df[col].apply(
        lambda x: le.transform([x])[0] if x in known else -1
    )
    n_unknown = (test_df[col + "_encoded"] == -1).sum()
    return le, n_unknown


def encode_dataset(train_df, test_df, sir_cols):
    print(f"\n--- Encoding: Age Group (ordinal) ---")
    encode_ordinal(train_df, test_df, "Age Group", age_order)

    print(f"--- Encoding: {len(sir_cols)} antibiotic S/I/R columns (ordinal) ---")
    for col in sir_cols:
        encode_ordinal(train_df, test_df, col, sir_order)

    print(f"--- Encoding: Species, Continent, Source_Category, Speciality, Gender (nominal) ---")
    nominal_cols = ["Species", "Continent", "Source_Category", "Speciality", "Gender"]
    for col in nominal_cols:
        le, n_unknown = encode_nominal_with_unknown(train_df, test_df, col)
        print(f"    {col}: {len(le.classes_)} categories learned, "
              f"{n_unknown} 'Unknown' in test ({n_unknown/len(test_df)*100:.3f}%)")

    return train_df, test_df


# ------------------------------------------------------------
# Apply to Model A (Carbapenem)
# ------------------------------------------------------------
sir_cols_carb = [c for c in approved_features_carbapenem if c.endswith("_I")]

print("=" * 70)
print("ENCODING — MODEL A (CARBAPENEM)")
print("=" * 70)
train_carb, test_carb = encode_dataset(train_carb, test_carb, sir_cols_carb)

# ------------------------------------------------------------
# Apply to Model B (Cephalosporin)
# ------------------------------------------------------------
sir_cols_ceph = [c for c in approved_features_cephalosporin if c.endswith("_I")]

print("\n" + "=" * 70)
print("ENCODING — MODEL B (CEPHALOSPORIN)")
print("=" * 70)
train_ceph, test_ceph = encode_dataset(train_ceph, test_ceph, sir_cols_ceph)

print("\n✅ Encoding complete for both models — fit on training data only.")
```

```python
# ============================================================
# CELL 18: DIAGNOSTICS — COMPUTED ON TRAINING DATA ONLY
# ============================================================

import numpy as np
from scipy.stats import chi2_contingency

def cramers_v(col1, col2, data):
    contingency = pd.crosstab(data[col1], data[col2])
    chi2 = chi2_contingency(contingency)[0]
    n = contingency.sum().sum()
    min_dim = min(contingency.shape) - 1
    return np.sqrt(chi2 / (n * min_dim)) if min_dim > 0 else np.nan

def run_diagnostics(train_df, feature_cols, target_col, model_name):
    print("=" * 70)
    print(f"DIAGNOSTICS — {model_name} (training data only)")
    print("=" * 70)

    counts = train_df[target_col].value_counts()
    ratio = counts[0] / counts[1]
    print(f"\nClass balance:")
    print(f"  Not Resistant: {counts[0]:,}")
    print(f"  Resistant:     {counts[1]:,}")
    print(f"  Ratio: 1 : {ratio:.2f}")

    encoded_cols = [c + "_encoded" for c in feature_cols if c + "_encoded" in train_df.columns]
    print(f"\nComputing Cramér's V across {len(encoded_cols)} encoded features...")

    corr = pd.DataFrame(index=encoded_cols, columns=encoded_cols, dtype=float)
    for c1 in encoded_cols:
        for c2 in encoded_cols:
            corr.loc[c1, c2] = round(cramers_v(c1, c2, train_df), 3)

    high_corr = []
    for c1 in encoded_cols:
        for c2 in encoded_cols:
            if c1 < c2 and corr.loc[c1, c2] >= 0.6:
                high_corr.append((c1, c2, corr.loc[c1, c2]))

    print(f"\nPairs with Cramér's V >= 0.6 (potential redundancy):")
    if high_corr:
        for c1, c2, v in high_corr:
            print(f"  {c1} <-> {c2}: {v}")
    else:
        print("  None found — no strong redundancy detected.")

    return corr, ratio

corr_carb, ratio_carb = run_diagnostics(train_carb, approved_features_carbapenem, "Carbapenem_Resistant", "MODEL A")
corr_ceph, ratio_ceph = run_diagnostics(train_ceph, approved_features_cephalosporin, "Cephalosporin_Resistant", "MODEL B")
```

```python
# ============================================================
# CELL 19: CORRELATION HEATMAP — VISUAL VERSION OF CELL 18
# ============================================================
# Builds on corr_carb and corr_ceph already computed in Cell 18.
# Two heatmaps, one per model, saved as PNG for the thesis.

import matplotlib.pyplot as plt
import seaborn as sns

# Safety check — confirm Cell 18 was run
required = ["corr_carb", "corr_ceph"]
missing = [r for r in required if r not in globals()]
if missing:
    raise ValueError(f"Missing from Cell 18: {missing}. Please run Cell 18 first.")


def plot_correlation_heatmap(corr_matrix, title, filename):
    """
    Plots a Cramér's V correlation heatmap. Cells >= 0.6 are
    outlined to make redundancy checks easy to spot visually.
    """
    corr_float = corr_matrix.astype(float)

    # Shorten column names for readability (strip "_I_encoded" etc.)
    short_labels = [
        c.replace("_encoded", "").replace("_I", "")
        for c in corr_float.columns
    ]
    corr_float.columns = short_labels
    corr_float.index = short_labels

    fig, ax = plt.subplots(figsize=(14, 12))

    sns.heatmap(
        corr_float,
        cmap="Reds",
        vmin=0, vmax=1,
        annot=True, fmt=".2f",
        annot_kws={"size": 7},
        linewidths=0.4, linecolor="white",
        cbar_kws={"label": "Cramér's V"},
        square=True,
        ax=ax
    )

    # Outline cells >= 0.6 to flag potential redundancy visually
    for i in range(corr_float.shape[0]):
        for j in range(corr_float.shape[1]):
            if i != j and corr_float.iloc[i, j] >= 0.6:
                ax.add_patch(
                    plt.Rectangle((j, i), 1, 1, fill=False,
                                  edgecolor="blue", lw=2)
                )

    ax.set_title(
        f"{title}\nCramér's V Correlation Among Encoded Features "
        f"(training data only)\nBlue outline = V ≥ 0.6",
        fontsize=13, fontweight="bold", pad=15
    )
    plt.xticks(rotation=45, ha="right", fontsize=8)
    plt.yticks(rotation=0, fontsize=8)
    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()
    print(f"✅ Saved: {filename}")


print("=" * 70)
print("CORRELATION HEATMAP — MODEL A (CARBAPENEM)")
print("=" * 70)
plot_correlation_heatmap(
    corr_carb,
    "Model A: Carbapenem Resistance — Feature Correlation",
    "Correlation_Heatmap_Model_A_Carbapenem.png"
)

print("\n" + "=" * 70)
print("CORRELATION HEATMAP — MODEL B (CEPHALOSPORIN)")
print("=" * 70)
plot_correlation_heatmap(
    corr_ceph,
    "Model B: Cephalosporin Resistance — Feature Correlation",
    "Correlation_Heatmap_Model_B_Cephalosporin.png"
)
```

```python
# ============================================================
# PRE-TRAINING VERIFICATION 
# ============================================================

def get_final_feature_columns(approved_features):
    """
    Returns the correct column name for each feature:
    - Year stays as "Year" (already numeric, never encoded)
    - Everything else uses its "_encoded" version
    """
    final_cols = []
    for f in approved_features:
        if f == "Year":
            final_cols.append("Year")
        else:
            final_cols.append(f + "_encoded")
    return final_cols


def verify_ready(X_train, X_test, y_train, y_test, model_name):
    print("=" * 70)
    print(f"VERIFICATION — {model_name}")
    print("=" * 70)

    print(f"\nX_train shape: {X_train.shape}")
    print(f"X_test shape:  {X_test.shape}")
    print(f"y_train shape: {y_train.shape}")
    print(f"y_test shape:  {y_test.shape}")

    print(f"\nAll X_train columns numeric? {all(X_train.dtypes.apply(lambda d: d.kind in 'if'))}")
    print(f"All X_test columns numeric?  {all(X_test.dtypes.apply(lambda d: d.kind in 'if'))}")

    print(f"\nMissing values in X_train: {X_train.isna().sum().sum()}")
    print(f"Missing values in X_test:  {X_test.isna().sum().sum()}")
    print(f"Missing values in y_train: {y_train.isna().sum()}")
    print(f"Missing values in y_test:  {y_test.isna().sum()}")

    print(f"\ny_train unique values: {sorted(y_train.unique())}")
    print(f"y_test unique values:  {sorted(y_test.unique())}")

    print("\nColumn list used as features:")
    for c in X_train.columns:
        print(f"  - {c}")

    all_checks_passed = (
        X_train.shape[1] == X_test.shape[1] and
        X_train.isna().sum().sum() == 0 and
        X_test.isna().sum().sum() == 0 and
        y_train.isna().sum() == 0 and
        y_test.isna().sum() == 0 and
        set(y_train.unique()).issubset({0.0, 1.0}) and
        set(y_test.unique()).issubset({0.0, 1.0})
    )

    print("\n" + ("🏆 READY FOR TRAINING" if all_checks_passed else "⚠️  NOT READY — see issues above"))
    return all_checks_passed


# ------------------------------------------------------------
# Model A (Carbapenem)
# ------------------------------------------------------------
feature_cols_carb = get_final_feature_columns(approved_features_carbapenem)

X_train_carb = train_carb[feature_cols_carb]
X_test_carb  = test_carb[feature_cols_carb]
y_train_carb = train_carb["Carbapenem_Resistant"]
y_test_carb  = test_carb["Carbapenem_Resistant"]

ready_A = verify_ready(X_train_carb, X_test_carb, y_train_carb, y_test_carb, "MODEL A (CARBAPENEM)")

# ------------------------------------------------------------
# Model B (Cephalosporin)
# ------------------------------------------------------------
feature_cols_ceph = get_final_feature_columns(approved_features_cephalosporin)

X_train_ceph = train_ceph[feature_cols_ceph]
X_test_ceph  = test_ceph[feature_cols_ceph]
y_train_ceph = train_ceph["Cephalosporin_Resistant"]
y_test_ceph  = test_ceph["Cephalosporin_Resistant"]

ready_B = verify_ready(X_train_ceph, X_test_ceph, y_train_ceph, y_test_ceph, "MODEL B (CEPHALOSPORIN)")

# ------------------------------------------------------------
# Final summary
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("FINAL SUMMARY")
print("=" * 70)
print(f"Model A ready: {'✅ Yes' if ready_A else '❌ No'}")
print(f"Model B ready: {'✅ Yes' if ready_B else '❌ No'}")
```

```python
# ============================================================
# CELL 21 (EXPANDED): MULTI-MODEL COMPARISON — MODEL A
# Now includes 9 models
# ============================================================

import time
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, average_precision_score, confusion_matrix,
    balanced_accuracy_score, matthews_corrcoef
)

try:
    from lightgbm import LGBMClassifier
    HAS_LGBM = True
except ImportError:
    HAS_LGBM = False
    print("⚠️  lightgbm not installed — skipped. (pip install lightgbm)")

try:
    from catboost import CatBoostClassifier
    HAS_CATBOOST = True
except ImportError:
    HAS_CATBOOST = False
    print("⚠️  catboost not installed — skipped. (pip install catboost)")

# ------------------------------------------------------------
# 0. Recalculate moderated_weight fresh (safe if kernel reset)
# ------------------------------------------------------------
n_not_resistant = (y_train_carb == 0).sum()
n_resistant = (y_train_carb == 1).sum()
full_ratio = n_not_resistant / n_resistant
moderated_weight = full_ratio / 2
print(f"Moderated weight: {moderated_weight:.2f}\n")

# ------------------------------------------------------------
# 1. Scaled versions for scale-sensitive models (fit on TRAIN only)
# ------------------------------------------------------------
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_carb)
X_test_scaled  = scaler.transform(X_test_carb)

# ------------------------------------------------------------
# 2. Define all models
# ------------------------------------------------------------
models = {
    "Logistic Regression": (LogisticRegression(max_iter=1000, class_weight="balanced",
                                                random_state=42), True),
    "KNN":                 (KNeighborsClassifier(n_neighbors=25), True),
    "Neural Network (MLP)": (MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=200,
                                            early_stopping=True, random_state=42), True),
    "Random Forest":       (RandomForestClassifier(n_estimators=200, class_weight="balanced",
                                                    n_jobs=-1, random_state=42), False),
    "Gradient Boosting":   (GradientBoostingClassifier(n_estimators=200, max_depth=4,
                                                        learning_rate=0.1, random_state=42), False),
    "AdaBoost":            (AdaBoostClassifier(n_estimators=200, learning_rate=0.5,
                                               random_state=42), False),
    "XGBoost":             (XGBClassifier(n_estimators=300, max_depth=6, learning_rate=0.1,
                                          scale_pos_weight=moderated_weight,
                                          eval_metric="auc", random_state=42), False),
}
if HAS_LGBM:
    models["LightGBM"] = (LGBMClassifier(n_estimators=300, max_depth=6, learning_rate=0.1,
                                         class_weight="balanced", random_state=42, verbose=-1), False)
if HAS_CATBOOST:
    models["CatBoost"] = (CatBoostClassifier(iterations=300, depth=6, learning_rate=0.1,
                                             auto_class_weights="Balanced",
                                             random_state=42, verbose=0), False)

# ------------------------------------------------------------
# 3. Train + evaluate each model
# ------------------------------------------------------------
results = []

for name, (model, needs_scaling) in models.items():
    print(f"Training {name}...")
    start = time.time()

    Xtr = X_train_scaled if needs_scaling else X_train_carb
    Xte = X_test_scaled  if needs_scaling else X_test_carb

    model.fit(Xtr, y_train_carb)
    y_pred = model.predict(Xte)
    y_proba = model.predict_proba(Xte)[:, 1]

    tn, fp, fn, tp = confusion_matrix(y_test_carb, y_pred).ravel()

    results.append({
        "Model": name,
        "ROC-AUC": round(roc_auc_score(y_test_carb, y_proba), 4),
        "PR-AUC": round(average_precision_score(y_test_carb, y_proba), 4),
        "Accuracy": round(accuracy_score(y_test_carb, y_pred), 4),
        "Precision": round(precision_score(y_test_carb, y_pred), 4),
        "Recall/Sens": round(tp/(tp+fn), 4),
        "Specificity": round(tn/(tn+fp), 4),
        "F1": round(f1_score(y_test_carb, y_pred), 4),
        "Bal.Acc": round(balanced_accuracy_score(y_test_carb, y_pred), 4),
        "MCC": round(matthews_corrcoef(y_test_carb, y_pred), 4),
        "Time(s)": round(time.time() - start, 1),
    })
    print(f"  ✅ done in {results[-1]['Time(s)']}s — ROC-AUC {results[-1]['ROC-AUC']}")

# ------------------------------------------------------------
# 4. Comparison table
# ------------------------------------------------------------
results_df = pd.DataFrame(results).sort_values("ROC-AUC", ascending=False).reset_index(drop=True)

print("\n" + "=" * 90)
print(f"MODEL COMPARISON — MODEL A (CARBAPENEM), TEST SET 2020–2024  |  {len(results_df)} models")
print("=" * 90)
display(results_df)
```

```python
# ============================================================
# CELL 26B: SHAP — CLEAN LABELS + IMPORTANCE TABLE + BAR CHART
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import shap

# ------------------------------------------------------------
# 1. Clean, human-readable feature name mapping
# ------------------------------------------------------------
def clean_feature_name(raw_name):
    name = raw_name.replace("_I_encoded", "").replace("_encoded", "")
    rename_map = {
        "Piperacillin tazobactam": "Piperacillin-Tazobactam",
        "Amoxycillin clavulanate": "Amoxicillin-Clavulanate",
        "Trimethoprim sulfa": "Trimethoprim-Sulfamethoxazole",
        "Ampicillin sulbactam": "Ampicillin-Sulbactam",
        "Age Group": "Age Group",
        "Source_Category": "Specimen Source",
        "Speciality": "Care Setting",
        "Species": "Bacterial Species",
        "Continent": "Geographic Region",
    }
    return rename_map.get(name, name)


def shap_importance_table(shap_values, feature_names, model_label):
    """Builds a clean, sorted importance table: mean |SHAP value| per feature."""
    mean_abs_shap = np.abs(shap_values).mean(axis=0)
    clean_names = [clean_feature_name(f) for f in feature_names]

    table = pd.DataFrame({
        "Feature": clean_names,
        "Mean |SHAP value|": mean_abs_shap
    }).sort_values("Mean |SHAP value|", ascending=False).reset_index(drop=True)

    table["Rank"] = table.index + 1
    table = table[["Rank", "Feature", "Mean |SHAP value|"]]
    table["Mean |SHAP value|"] = table["Mean |SHAP value|"].round(4)

    print(f"\n{'='*70}")
    print(f"SHAP FEATURE IMPORTANCE TABLE — {model_label}")
    print(f"{'='*70}")
    display(table)
    return table


def plot_clean_bar(table, model_label, filename, top_n=15):
    """Clean horizontal bar chart, top N features, readable labels."""
    plot_data = table.head(top_n).iloc[::-1]  # reverse so #1 is on top

    fig, ax = plt.subplots(figsize=(9, 7))
    ax.barh(plot_data["Feature"], plot_data["Mean |SHAP value|"], color="#C00000")
    ax.set_xlabel("Mean |SHAP value| (average impact on model output)", fontsize=11)
    ax.set_title(f"{model_label} — Feature Importance (SHAP)", fontsize=13, fontweight="bold")
    ax.tick_params(axis="y", labelsize=10)
    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()


def plot_clean_beeswarm(shap_values, X_sample, feature_names, model_label, filename):
    """Beeswarm plot with clean, readable feature labels."""
    X_display = X_sample.copy()
    X_display.columns = [clean_feature_name(c) for c in feature_names]

    plt.figure(figsize=(9, 8))
    shap.summary_plot(shap_values, X_display, show=False, max_display=15)
    plt.title(f"{model_label} — SHAP Summary (Direction & Magnitude)",
              fontsize=13, fontweight="bold")
    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()


# ------------------------------------------------------------
# 2. Run for MODEL A (XGBoost — Carbapenem)
# ------------------------------------------------------------
table_A = shap_importance_table(shap_values_A, sample_A.columns.tolist(), "MODEL A (Carbapenem Resistance)")
plot_clean_bar(table_A, "Model A (Carbapenem Resistance)", "SHAP_BarChart_Model_A.png")
plot_clean_beeswarm(shap_values_A, sample_A, sample_A.columns.tolist(), "Model A (Carbapenem Resistance)", "SHAP_Beeswarm_Model_A_Clean.png")

# ------------------------------------------------------------
# 3. Run for MODEL B (CatBoost — Cephalosporin)
# ------------------------------------------------------------
table_B = shap_importance_table(shap_values_B, sample_B.columns.tolist(), "MODEL B (Cephalosporin Resistance)")
plot_clean_bar(table_B, "Model B (Cephalosporin Resistance)", "SHAP_BarChart_Model_B.png")
plot_clean_beeswarm(shap_values_B, sample_B, sample_B.columns.tolist(), "Model B (Cephalosporin Resistance)", "SHAP_Beeswarm_Model_B_Clean.png")
```
