```python
# ============================================================
# SO1 — BLOCK 2
# CREATE BINARY ANTIMICROBIAL RESISTANCE OUTCOMES
# WHO PRIORITY GRAM-NEGATIVE PATHOGENS
# ============================================================

import numpy as np
import pandas as pd

df_gn = df_gn.copy()

print("=" * 80)
print("CREATING BINARY RESISTANCE OUTCOMES")
print("=" * 80)

# ============================================================
# PHENOTYPIC RESISTANCE DEFINITION
# ------------------------------------------------------------
# An isolate is considered resistant if it is interpreted as
# "Resistant" to ANY antimicrobial within the antibiotic class.
#
# Clinical interpretation follows CLSI/EUCAST S/I/R categories.
# ============================================================

# ============================================================
# CARBAPENEMS
# ============================================================

carbapenem_cols = [
    "Meropenem_I",
    "Imipenem_I",
    "Doripenem_I",
    "Ertapenem_I"
]

# ============================================================
# THIRD-GENERATION CEPHALOSPORINS
#
# (Expanded according to supervisor recommendation)
# ============================================================

cephalosporin_cols = [
    "Ceftriaxone_I",
    "Ceftazidime_I",
    "Cefpodoxime_I",
    "Cefixime_I",
    "Ceftibuten_I"
]

# ============================================================
# KEEP ONLY VARIABLES AVAILABLE IN DATASET
# ============================================================

carbapenem_cols = [
    c for c in carbapenem_cols
    if c in df_gn.columns
]

cephalosporin_cols = [
    c for c in cephalosporin_cols
    if c in df_gn.columns
]

print("\nCarbapenem drugs included:")
for c in carbapenem_cols:
    print("  •", c)

print("\nThird-generation cephalosporins included:")
for c in cephalosporin_cols:
    print("  •", c)

# ============================================================
# CREATE BINARY RESISTANCE FUNCTION
# ============================================================

def create_binary_resistance(df, antibiotic_columns):
    """
    Create binary resistance outcome.

    1 = Resistant to at least one drug in the class
    0 = No resistant result within the class
    NaN = No antimicrobial tested
    """

    resistant = (
        df[antibiotic_columns]
        .eq("Resistant")
        .any(axis=1)
        .astype(float)
    )

    resistant[
        df[antibiotic_columns].isna().all(axis=1)
    ] = np.nan

    return resistant

# ============================================================
# CREATE RESISTANCE OUTCOMES
# ============================================================

df_gn["Carbapenem_Resistant"] = create_binary_resistance(
    df_gn,
    carbapenem_cols
)

df_gn["Cephalosporin_Resistant"] = create_binary_resistance(
    df_gn,
    cephalosporin_cols
)

# ============================================================
# NUMBER OF DRUGS TESTED PER ISOLATE
# ============================================================

df_gn["N_Carbapenems_Tested"] = (
    df_gn[carbapenem_cols]
    .notna()
    .sum(axis=1)
)

df_gn["N_Cephalosporins_Tested"] = (
    df_gn[cephalosporin_cols]
    .notna()
    .sum(axis=1)
)

# ============================================================
# SUMMARY FUNCTION
# ============================================================

def summarize_resistance(df, variable, tested_variable):

    tested = (df[tested_variable] > 0).sum()

    resistant = (df[variable] == 1).sum()

    susceptible = (df[variable] == 0).sum()

    missing = df[variable].isna().sum()

    rate = resistant / tested * 100 if tested > 0 else np.nan

    print("\n" + "=" * 70)
    print(variable.upper())
    print("=" * 70)
    print(f"Total isolates               : {len(df):,}")
    print(f"Isolates tested              : {tested:,}")
    print(f"Resistant isolates           : {resistant:,}")
    print(f"Non-resistant isolates       : {susceptible:,}")
    print(f"Missing phenotype            : {missing:,}")
    print(f"Resistance rate (%)          : {rate:.2f}")

# ============================================================
# PRINT PUBLICATION-READY SUMMARY
# ============================================================

summarize_resistance(
    df_gn,
    "Carbapenem_Resistant",
    "N_Carbapenems_Tested"
)

summarize_resistance(
    df_gn,
    "Cephalosporin_Resistant",
    "N_Cephalosporins_Tested"
)

# ============================================================
# SAVE DATASET (OPTIONAL)
# ============================================================

# df_gn.to_csv(
#     "WHO_Priority_GramNegative_Resistance_Outcomes.csv",
#     index=False
# )

print("\n" + "=" * 80)
print("BINARY RESISTANCE VARIABLES CREATED SUCCESSFULLY")
print("=" * 80)
print("Variables added:")
print("  • Carbapenem_Resistant")
print("  • Cephalosporin_Resistant")
print("  • N_Carbapenems_Tested")
print("  • N_Cephalosporins_Tested")
```

```python
# ============================================================
# SO1 — BLOCK 3
# PUBLICATION-GRADE ANNUAL AMR SURVEILLANCE TRENDS
# ============================================================

import pandas as pd
import numpy as np


print("="*90)
print("SO1 — BLOCK 3: ANNUAL ANTIMICROBIAL RESISTANCE TRENDS")
print("="*90)


# ============================================================
# FUNCTION: WILSON CI
# ============================================================

def wilson_ci(x, n, z=1.96):

    if n == 0:
        return np.nan, np.nan

    p = x/n

    denominator = 1 + z**2/n

    center = (
        p + z*z/(2*n)
    ) / denominator

    margin = (
        z *
        np.sqrt(
            (p*(1-p)/n)
            +
            (z*z/(4*n*n))
        )
    ) / denominator


    return (
        max(0, center-margin),
        min(1, center+margin)
    )



# ============================================================
# FUNCTION: ANNUAL RATE CALCULATION
# ============================================================

def calculate_amr_trend(
        data,
        outcome,
        group=None):


    results=[]


    if group is None:

        datasets=[
            ("Overall", data)
        ]

    else:

        if group not in data.columns:

            print(
                f"WARNING: {group} not found. Skipping."
            )

            return pd.DataFrame()


        datasets=data.groupby(group)



    for name,df_temp in datasets:


        for year in sorted(df_temp["Year"].dropna().unique()):


            year_df=df_temp[
                df_temp["Year"]==year
            ][outcome].dropna()


            n=len(year_df)


            if n==0:
                continue


            resistant=int(year_df.sum())


            rate=resistant/n


            lower,upper=wilson_ci(
                resistant,
                n
            )


            results.append({

                "Group":name,

                "Year":year,

                "N_Tested":n,

                "N_Resistant":resistant,

                "Resistance_Rate_%":
                    round(rate*100,2),

                "CI_Lower_%":
                    round(lower*100,2),

                "CI_Upper_%":
                    round(upper*100,2)

            })


    return pd.DataFrame(results)



# ============================================================
# 1. GLOBAL TEMPORAL TRENDS
# ============================================================


annual_carb = calculate_amr_trend(
    df_gn,
    "Carbapenem_Resistant"
)


annual_ceph = calculate_amr_trend(
    df_gn,
    "Cephalosporin_Resistant"
)



print("\nGLOBAL CARBAPENEM TREND")
display(annual_carb)


print("\nGLOBAL CEPHALOSPORIN TREND")
display(annual_ceph)



# ============================================================
# 2. SPECIES LEVEL TRENDS
# ============================================================


species_carb = calculate_amr_trend(
    df_gn,
    "Carbapenem_Resistant",
    group="Species"
)


species_ceph = calculate_amr_trend(
    df_gn,
    "Cephalosporin_Resistant",
    group="Species"
)


print("\nSpecies-level carbapenem trends created")
print(species_carb.shape)


print("\nSpecies-level cephalosporin trends created")
print(species_ceph.shape)



# ============================================================
# 3. PATHOGEN GROUP TRENDS
# ============================================================


group_carb = calculate_amr_trend(
    df_gn,
    "Carbapenem_Resistant",
    group="Pathogen_Group"
)


group_ceph = calculate_amr_trend(
    df_gn,
    "Cephalosporin_Resistant",
    group="Pathogen_Group"
)



# ============================================================
# 4. WHO REGION (ONLY IF AVAILABLE)
# ============================================================


if "WHO_Region" in df_gn.columns:


    regional_carb = calculate_amr_trend(
        df_gn,
        "Carbapenem_Resistant",
        group="WHO_Region"
    )


    regional_ceph = calculate_amr_trend(
        df_gn,
        "Cephalosporin_Resistant",
        group="WHO_Region"
    )


    print("\nWHO regional trends generated")


else:


    print(
        "\nWHO_Region column not available."
    )

    print(
        "Regional analysis will be performed after WHO region assignment."
    )


# ============================================================
# SAVE OUTPUTS
# ============================================================


annual_carb.to_csv(
    "SO1_Global_Carbapenem_Annual_Trends.csv",
    index=False
)


annual_ceph.to_csv(
    "SO1_Global_Cephalosporin_Annual_Trends.csv",
    index=False
)


species_carb.to_csv(
    "SO1_Species_Carbapenem_Trends.csv",
    index=False
)


species_ceph.to_csv(
    "SO1_Species_Cephalosporin_Trends.csv",
    index=False
)


group_carb.to_csv(
    "SO1_Pathogen_Group_Carbapenem_Trends.csv",
    index=False
)


group_ceph.to_csv(
    "SO1_Pathogen_Group_Cephalosporin_Trends.csv",
    index=False
)



print("\n")
print("="*90)
print("BLOCK 3 COMPLETED SUCCESSFULLY")
print("="*90)
```

```python
# ============================================================
# SO1 — BLOCK 5 (CORRECTED): COMPREHENSIVE STATISTICAL TREND ANALYSIS
# ============================================================
# REQUIRED HELPER FUNCTION — MUST RUN BEFORE BLOCK 5
# (Calculates annual resistance rate + Wilson 95% CI for any subgroup)
# ============================================================
def calculate_annual_rates(df, outcome_col):
    results = []
    for year in sorted(df["Year"].dropna().unique()):
        year_df = df[df["Year"] == year][outcome_col].dropna()
        n = len(year_df)
        if n == 0:
            continue
        resistant = int(year_df.sum())
        rate = resistant / n
        z = 1.96
        denom = 1 + (z**2 / n)
        center = (rate + (z**2 / (2*n))) / denom
        margin = (z * np.sqrt((rate*(1-rate)/n) + (z**2/(4*n**2)))) / denom
        results.append({
            "Year": year, "N_Tested": n, "N_Resistant": resistant,
            "Resistance_Rate_%": round(rate*100, 2),
            "CI_Lower_%": round(max(0, center-margin)*100, 2),
            "CI_Upper_%": round(min(1, center+margin)*100, 2)
        })
    return pd.DataFrame(results)
# ============================================================
#
# CORRECTIONS APPLIED (per supervisor review):
# 1. Fixed GLS function — conf_int() array indexing bug
# 2. Section 2 now analyzes Enterobacterales, Pseudomonas,
#    and Acinetobacter SEPARATELY (matches Pathogen_Group)
# 3. Trend interpretation now checks significance —
#    reports "No significant trend" when CI crosses zero
#    or p >= 0.05, instead of mislabeling as Increasing/Decreasing
#
# ============================================================

import numpy as np
import pandas as pd
from scipy import stats
import statsmodels.api as sm
import warnings
warnings.filterwarnings("ignore")

# ============================================================
# SAFETY CHECK
# ============================================================

required = [
    "df_gn", "annual_carb", "annual_ceph",
    "carbapenem_cols", "cephalosporin_cols",
    "calculate_annual_rates", "Pathogen_Group"
    if "Pathogen_Group" in df_gn.columns else "df_gn"
]
missing_obj = [
    r for r in required
    if r not in globals() and r != "Pathogen_Group"
]
if missing_obj:
    raise ValueError(
        f"Missing objects from previous blocks: {missing_obj}\n"
        f"Please run Blocks 1–4 first."
    )

if "Pathogen_Group" not in df_gn.columns:
    raise ValueError(
        "Pathogen_Group column not found in df_gn. "
        "Please confirm Block 1 created this classification "
        "(Enterobacterales / Pseudomonas / Acinetobacter)."
    )

print("=" * 80)
print("SO1 — BLOCK 5 (CORRECTED): COMPREHENSIVE STATISTICAL TREND ANALYSIS")
print("=" * 80)
print(f"Dataset: {len(df_gn):,} isolates | "
      f"{df_gn['Year'].nunique()} years | "
      f"{df_gn['Country'].nunique()} countries")
print(f"Pathogen groups: {df_gn['Pathogen_Group'].unique().tolist()}")


# ============================================================
# FUNCTION 1: COCHRAN-ARMITAGE TREND TEST (unchanged — correct)
# ============================================================

def cochran_armitage_test(df, outcome_col):

    yearly = (
        df.groupby("Year")[outcome_col]
        .agg(["sum", "count"])
        .reset_index()
        .dropna()
    )

    t = np.arange(len(yearly))
    y = yearly["sum"].values.astype(float)
    n = yearly["count"].values.astype(float)

    mask = n > 0
    t, y, n = t[mask], y[mask], n[mask]

    p_overall = y.sum() / n.sum()
    expected  = n * p_overall
    t_bar     = np.average(t, weights=n)

    numerator   = np.sum(t * (y - expected))
    denominator = np.sqrt(
        p_overall * (1 - p_overall) *
        np.sum(n * (t - t_bar)**2)
    )

    if denominator == 0:
        return None

    Z       = numerator / denominator
    p_value = 2 * (1 - stats.norm.cdf(abs(Z)))

    return {
        "Test":        "Cochran-Armitage",
        "Z_statistic": round(Z, 4),
        "P_value":     p_value,
        "Significant": p_value < 0.05
    }


# ============================================================
# FUNCTION 2: MANN-KENDALL TREND TEST (unchanged — correct)
# ============================================================

def mann_kendall_test(series):

    x = np.array(series)
    n = len(x)
    S = 0

    for i in range(n - 1):
        for j in range(i + 1, n):
            S += np.sign(x[j] - x[i])

    var_s = (n * (n - 1) * (2*n + 5)) / 18

    if S > 0:
        Z = (S - 1) / np.sqrt(var_s)
    elif S < 0:
        Z = (S + 1) / np.sqrt(var_s)
    else:
        Z = 0.0

    p_value = 2 * (1 - stats.norm.cdf(abs(Z)))

    return {
        "Test":        "Mann-Kendall",
        "Z_statistic": round(Z, 4),
        "P_value":     p_value,
        "Significant": p_value < 0.05
    }


# ============================================================
# FUNCTION 3: APC WITH CORRECTED SIGNIFICANCE-AWARE LABELING
# ============================================================
# FIX APPLIED: Trend label now depends on statistical
# significance, not just the sign of APC.
#
# Rule:
#   If p >= 0.05 OR the 95% CI crosses zero
#   -> "No significant trend"
#   Else
#   -> "Increasing ↑" or "Decreasing ↓" based on sign
# ============================================================

def calculate_apc(annual_df, rate_col="Resistance_Rate_%"):

    df_c = annual_df[["Year", rate_col]].dropna()
    df_c = df_c[df_c[rate_col] > 0].copy()

    if len(df_c) < 5:
        return None

    years     = df_c["Year"].values.astype(float)
    rates     = df_c[rate_col].values
    log_rates = np.log(rates / 100)

    slope, intercept, r, p, se = stats.linregress(years, log_rates)

    apc   = (np.exp(slope) - 1) * 100
    ci_l  = (np.exp(slope - 1.96*se) - 1) * 100
    ci_u  = (np.exp(slope + 1.96*se) - 1) * 100

    # ---------------------------------------------------
    # CORRECTED TREND LOGIC — significance-aware
    # ---------------------------------------------------
    ci_crosses_zero = (ci_l < 0 < ci_u)
    not_significant  = (p >= 0.05) or ci_crosses_zero

    if not_significant:
        trend_label = "No significant trend"
    elif apc > 0:
        trend_label = "Increasing ↑"
    else:
        trend_label = "Decreasing ↓"

    return {
        "APC_%":       round(apc, 3),
        "CI_Lower_%":  round(ci_l, 3),
        "CI_Upper_%":  round(ci_u, 3),
        "R_squared":   round(r**2, 4),
        "P_value":     p,
        "Significant": not not_significant,
        "Trend":       trend_label
    }


# ============================================================
# FUNCTION 4: WEIGHTED GLS REGRESSION — BUG FIXED
# ============================================================
# FIX APPLIED: conf_int() output is coerced to a numpy
# array with np.asarray() before indexing, which resolves
# the failure that previously caused GLS to silently
# return "N/A" for every row.
# ============================================================

def weighted_gls_trend(annual_df, rate_col="Resistance_Rate_%"):

    df_c = annual_df[
        ["Year", rate_col, "N_Tested"]
    ].dropna()
    df_c = df_c[df_c[rate_col] > 0].copy()

    if len(df_c) < 5:
        return None

    years   = df_c["Year"].values.astype(float)
    rates   = df_c[rate_col].values
    weights = df_c["N_Tested"].values.astype(float)

    years_c = years - years.mean()
    X = sm.add_constant(years_c)

    try:
        model = sm.WLS(rates, X, weights=weights).fit()

        slope = float(model.params[1])
        p     = float(model.pvalues[1])
        r2    = float(model.rsquared)

        # --- FIX: force conf_int() result into a numpy array ---
        ci_array = np.asarray(model.conf_int(alpha=0.05))
        ci_l = float(ci_array[1, 0])
        ci_u = float(ci_array[1, 1])

        ci_crosses_zero = (ci_l < 0 < ci_u)
        not_significant  = (p >= 0.05) or ci_crosses_zero

        if not_significant:
            trend_label = "No significant trend"
        elif slope > 0:
            trend_label = "Increasing ↑"
        else:
            trend_label = "Decreasing ↓"

        return {
            "Method":         "Weighted GLS",
            "Slope_per_year": round(slope, 4),
            "CI_Lower":       round(ci_l, 4),
            "CI_Upper":       round(ci_u, 4),
            "R_squared":      round(r2, 4),
            "P_value":        p,
            "Significant":    not not_significant,
            "Trend":          trend_label,
            "Interpretation": (
                f"Resistance changes by "
                f"{slope:+.2f} percentage points per year "
                f"(95% CI: {ci_l:.2f} to {ci_u:.2f})"
            )
        }

    except Exception as e:
        print(f"    ⚠️  GLS error: {e}")
        return None


def format_p(p):
    return "< 0.001" if p < 0.001 else f"{p:.4f}"


# ============================================================
# SECTION 1: OVERALL STATISTICAL TESTS
# ============================================================

print("\n")
print("=" * 80)
print("SECTION 1: OVERALL RESISTANCE TREND STATISTICS")
print("=" * 80)

results_overall = []

for ab_label, annual_df, outcome_col in [
    ("Carbapenem", annual_carb, "Carbapenem_Resistant"),
    ("Third-Generation Cephalosporin", annual_ceph, "Cephalosporin_Resistant")
]:
    print(f"\n{'─'*70}")
    print(f"  {ab_label.upper()} RESISTANCE")
    print(f"{'─'*70}")

    ca  = cochran_armitage_test(df_gn, outcome_col)
    mk_series = df_gn.groupby("Year")[outcome_col].mean().values
    mk  = mann_kendall_test(mk_series)
    apc = calculate_apc(annual_df, "Resistance_Rate_%")
    gls = weighted_gls_trend(annual_df, "Resistance_Rate_%")

    first = annual_df.iloc[0]
    last  = annual_df.iloc[-1]

    print(f"\n  Resistance rate {int(first['Year'])}: "
          f"{first['Resistance_Rate_%']:.1f}% "
          f"(95% CI: {first['CI_Lower_%']:.1f}–{first['CI_Upper_%']:.1f}%)")
    print(f"  Resistance rate {int(last['Year'])}: "
          f"{last['Resistance_Rate_%']:.1f}% "
          f"(95% CI: {last['CI_Lower_%']:.1f}–{last['CI_Upper_%']:.1f}%)")

    print(f"\n  Cochran-Armitage Test:")
    print(f"    Z = {ca['Z_statistic']}, p {format_p(ca['P_value'])}, "
          f"Significant: {'Yes' if ca['Significant'] else 'No'}")

    print(f"\n  Mann-Kendall Test:")
    print(f"    Z = {mk['Z_statistic']}, p {format_p(mk['P_value'])}, "
          f"Significant: {'Yes' if mk['Significant'] else 'No'}")

    if apc:
        print(f"\n  Annual Percentage Change (APC):")
        print(f"    APC = {apc['APC_%']}% "
              f"(95% CI: {apc['CI_Lower_%']}–{apc['CI_Upper_%']}%), "
              f"R² = {apc['R_squared']}, p {format_p(apc['P_value'])}")
        print(f"    Interpretation: {apc['Trend']}")

    if gls:
        print(f"\n  Weighted GLS Regression:")
        print(f"    {gls['Interpretation']}, R² = {gls['R_squared']}, "
              f"p {format_p(gls['P_value'])}")
        print(f"    Interpretation: {gls['Trend']}")
    else:
        print(f"\n  Weighted GLS Regression: FAILED — insufficient data")

    results_overall.append({
        "Antibiotic Class":  ab_label,
        "Rate 2004 %":       f"{first['Resistance_Rate_%']:.1f}",
        "Rate 2024 %":       f"{last['Resistance_Rate_%']:.1f}",
        "CA Z-stat":         ca["Z_statistic"],
        "CA P-value":        format_p(ca["P_value"]),
        "MK Z-stat":         mk["Z_statistic"],
        "MK P-value":        format_p(mk["P_value"]),
        "APC %":             apc["APC_%"] if apc else "N/A",
        "APC 95% CI":        f"{apc['CI_Lower_%']}–{apc['CI_Upper_%']}" if apc else "N/A",
        "APC R²":            apc["R_squared"] if apc else "N/A",
        "APC P-value":       format_p(apc["P_value"]) if apc else "N/A",
        "GLS Slope/yr":      gls["Slope_per_year"] if gls else "N/A",
        "GLS 95% CI":        f"{gls['CI_Lower']}–{gls['CI_Upper']}" if gls else "N/A",
        "GLS P-value":       format_p(gls["P_value"]) if gls else "N/A",
        "Overall Trend":     apc["Trend"] if apc else "N/A"
    })

overall_df = pd.DataFrame(results_overall)
print("\n")
print("=" * 80)
print("TABLE 1: OVERALL STATISTICAL SUMMARY (CORRECTED)")
print("=" * 80)
display(overall_df)
overall_df.to_csv("SO1_Block5_Overall_Statistics_CORRECTED.csv", index=False)


# ============================================================
# SECTION 2 (CORRECTED): STRATIFIED APC BY THREE PATHOGEN GROUPS
# ============================================================
# FIX APPLIED: Now loops over Enterobacterales, Pseudomonas,
# and Acinetobacter SEPARATELY — matching the Pathogen_Group
# classification created in Block 1, instead of the outdated
# binary Enterobacterales / Non-Enterobacterales split.
# ============================================================

print("\n")
print("=" * 80)
print("SECTION 2 (CORRECTED): STRATIFIED APC BY PATHOGEN GROUP")
print("=" * 80)

pathogen_groups_present = df_gn["Pathogen_Group"].dropna().unique().tolist()
print(f"Pathogen groups detected in dataset: {pathogen_groups_present}")

stratified_rows = []

for group in pathogen_groups_present:

    group_df = df_gn[df_gn["Pathogen_Group"] == group]
    n_group  = len(group_df)

    if n_group < 500:
        print(f"  ⚠️  {group}: only {n_group} isolates — "
              f"results may be unstable")

    for ab_label, outcome_col in [
        ("Carbapenem",              "Carbapenem_Resistant"),
        ("Third-Gen Cephalosporin", "Cephalosporin_Resistant")
    ]:
        grp_annual = calculate_annual_rates(group_df, outcome_col)

        if len(grp_annual) < 5:
            continue

        apc = calculate_apc(grp_annual, "Resistance_Rate_%")
        ca  = cochran_armitage_test(group_df, outcome_col)
        mk_s = group_df.groupby("Year")[outcome_col].mean().values
        mk   = mann_kendall_test(mk_s)

        if not apc or not ca:
            continue

        first = grp_annual.iloc[0]
        last  = grp_annual.iloc[-1]

        stratified_rows.append({
            "Pathogen Group":       group,
            "Antibiotic Class":     ab_label,
            "N Isolates":           int(group_df[outcome_col].notna().sum()),
            "Rate 2004 % (95% CI)": (
                f"{first['Resistance_Rate_%']:.1f} "
                f"({first['CI_Lower_%']:.1f}–{first['CI_Upper_%']:.1f})"
            ),
            "Rate 2024 % (95% CI)": (
                f"{last['Resistance_Rate_%']:.1f} "
                f"({last['CI_Lower_%']:.1f}–{last['CI_Upper_%']:.1f})"
            ),
            "APC % (95% CI)": (
                f"{apc['APC_%']} ({apc['CI_Lower_%']}–{apc['CI_Upper_%']})"
            ),
            "APC R²":     apc["R_squared"],
            "CA P-value": format_p(ca["P_value"]),
            "MK P-value": format_p(mk["P_value"]),
            "Trend":      apc["Trend"]     # significance-aware label
        })

stratified_df = pd.DataFrame(stratified_rows)

print("\n")
print("TABLE 2 (CORRECTED): STRATIFIED BY PATHOGEN GROUP")
print("(Enterobacterales / Pseudomonas / Acinetobacter analyzed separately)")
print("=" * 80)
display(stratified_df)
stratified_df.to_csv(
    "SO1_Block5_Stratified_Statistics_CORRECTED.csv", index=False
)


# ============================================================
# SECTION 3: SPECIES-LEVEL STATISTICAL SUMMARY (CORRECTED LABELS)
# ============================================================

print("\n")
print("=" * 80)
print("SECTION 3: SPECIES-LEVEL STATISTICAL SUMMARY (CORRECTED)")
print("=" * 80)

key_species = [
    "Escherichia coli",
    "Klebsiella pneumoniae",
    "Pseudomonas aeruginosa",
    "Acinetobacter baumannii",
    "Enterobacter cloacae",
    "Proteus mirabilis",
    "Serratia marcescens"
]

species_rows = []

for species in key_species:

    sp_df = df_gn[df_gn["Species"] == species]

    if len(sp_df) < 100:
        continue

    for ab_label, outcome_col in [
        ("Carbapenem",       "Carbapenem_Resistant"),
        ("3G Cephalosporin", "Cephalosporin_Resistant")
    ]:
        sp_annual = calculate_annual_rates(sp_df, outcome_col)

        if len(sp_annual) < 5:
            continue

        apc = calculate_apc(sp_annual, "Resistance_Rate_%")
        ca  = cochran_armitage_test(sp_df, outcome_col)

        if not apc or not ca:
            continue

        first = sp_annual.iloc[0]
        last  = sp_annual.iloc[-1]

        species_rows.append({
            "Species":              species,
            "Antibiotic":           ab_label,
            "N Tested":             int(sp_df[outcome_col].notna().sum()),
            "Rate 2004 % (95% CI)": (
                f"{first['Resistance_Rate_%']:.1f} "
                f"({first['CI_Lower_%']:.1f}–{first['CI_Upper_%']:.1f})"
            ),
            "Rate 2024 % (95% CI)": (
                f"{last['Resistance_Rate_%']:.1f} "
                f"({last['CI_Lower_%']:.1f}–{last['CI_Upper_%']:.1f})"
            ),
            "APC % (95% CI)": (
                f"{apc['APC_%']} ({apc['CI_Lower_%']}–{apc['CI_Upper_%']})"
            ),
            "R²":       apc["R_squared"],
            "P-value":  format_p(apc["P_value"]),
            "Trend":    apc["Trend"]    # now significance-aware
        })

species_df = pd.DataFrame(species_rows)

print("\n")
print("TABLE 3 (CORRECTED): SPECIES-LEVEL STATISTICAL SUMMARY")
print("=" * 80)
display(species_df)
species_df.to_csv(
    "SO1_Block5_Species_Statistics_CORRECTED.csv", index=False
)


# ============================================================
# VERIFICATION: CONFIRM GLS FIX WORKED
# ============================================================

print("\n")
print("=" * 80)
print("VERIFICATION: GLS FIX CHECK")
print("=" * 80)

gls_working = overall_df["GLS Slope/yr"].apply(
    lambda x: x != "N/A"
).all()

if gls_working:
    print("✅ GLS regression now produces valid results for all rows.")
else:
    print("⚠️  GLS still returning N/A for some rows — review sample sizes.")


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 80)
print("BLOCK 5 (CORRECTED) COMPLETE — FILES SAVED")
print("=" * 80)
print("  📄 SO1_Block5_Overall_Statistics_CORRECTED.csv")
print("  📄 SO1_Block5_Stratified_Statistics_CORRECTED.csv")
print("  📄 SO1_Block5_Species_Statistics_CORRECTED.csv")
print("\n  Corrections applied:")
print("  ✅ GLS bug fixed (np.asarray on conf_int)")
print("  ✅ Section 2 now separates Enterobacterales / Pseudomonas / Acinetobacter")
print("  ✅ Trend labels now significance-aware ('No significant trend' added)")
print("=" * 80)
```

```python
# ============================================================
# ============================================================
# SO1 — BLOCK 6: SIGNIFICANCE-AWARE TEMPORAL VISUALIZATION
# ============================================================
#
# Purpose: Visualize the CORRECTED Block 5 results
# - Species-level temporal trends (with CI bands)
# - Pathogen-group trends (Enterobacterales/Pseudomonas/Acinetobacter)
# - Forest plots showing APC + 95% CI, colored by significance
# - Joinpoint regression re-run on corrected 589,607-isolate dataset
#
# Design principle: significant trends are shown in bold color;
# non-significant trends are shown in gray — so the reader
# never mistakes a non-significant APC for a real trend.
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from scipy import stats
import warnings
warnings.filterwarnings("ignore")

# ============================================================
# SAFETY CHECK
# ============================================================

required = [
    "df_gn", "annual_carb", "annual_ceph",
    "calculate_annual_rates", "calculate_apc",
    "stratified_df", "species_df"
]
missing_obj = [r for r in required if r not in globals()]
if missing_obj:
    raise ValueError(f"Missing from Block 5: {missing_obj}")

print("=" * 80)
print("SO1 — BLOCK 6: SIGNIFICANCE-AWARE TEMPORAL VISUALIZATION")
print("=" * 80)


# ============================================================
# SECTION 1: PATHOGEN-GROUP TEMPORAL TRENDS (3 GROUPS)
# ============================================================

group_colors = {
    "Enterobacterales": "#1F77B4",
    "Pseudomonas":       "#2CA02C",
    "Acinetobacter":     "#D62728"
}

fig, axes = plt.subplots(1, 2, figsize=(18, 7))

group_annual_store = {}   # keep for joinpoint section later

for group in df_gn["Pathogen_Group"].dropna().unique():

    group_df = df_gn[df_gn["Pathogen_Group"] == group]
    color = group_colors.get(group, "gray")

    carb_annual = calculate_annual_rates(group_df, "Carbapenem_Resistant")
    ceph_annual = calculate_annual_rates(group_df, "Cephalosporin_Resistant")

    group_annual_store[group] = {
        "carb": carb_annual, "ceph": ceph_annual
    }

    axes[0].plot(
        carb_annual["Year"], carb_annual["Resistance_Rate_%"],
        marker="o", color=color, linewidth=2.5, markersize=5,
        label=group
    )
    axes[0].fill_between(
        carb_annual["Year"],
        carb_annual["CI_Lower_%"], carb_annual["CI_Upper_%"],
        alpha=0.12, color=color
    )

    axes[1].plot(
        ceph_annual["Year"], ceph_annual["Resistance_Rate_%"],
        marker="s", color=color, linewidth=2.5, markersize=5,
        label=group
    )
    axes[1].fill_between(
        ceph_annual["Year"],
        ceph_annual["CI_Lower_%"], ceph_annual["CI_Upper_%"],
        alpha=0.12, color=color
    )

axes[0].set_title("A. Carbapenem Resistance by Pathogen Group", fontweight="bold")
axes[1].set_title("B. 3G Cephalosporin Resistance by Pathogen Group", fontweight="bold")

for ax in axes:
    ax.set_xlabel("Year")
    ax.set_ylabel("Resistance Rate (%)")
    ax.yaxis.set_major_formatter(mticker.PercentFormatter())
    ax.xaxis.set_major_locator(mticker.MultipleLocator(2))
    ax.legend(loc="upper left", frameon=False)
    ax.grid(alpha=0.3)
    ax.set_xlim(2004, 2024)

plt.suptitle(
    "Temporal Trends by Pathogen Group\n"
    "Enterobacterales vs. Pseudomonas vs. Acinetobacter — ATLAS 2004–2024",
    fontsize=14, fontweight="bold", y=1.03
)
plt.tight_layout()
plt.savefig("SO1_Block6_Fig1_PathogenGroup_Trends.png", dpi=300, bbox_inches="tight")
plt.show()
print("✅ Figure 1 saved: pathogen-group temporal trends")


# ============================================================
# SECTION 2: SIGNIFICANCE-AWARE FOREST PLOT — SPECIES LEVEL
# ============================================================
# Bold color = statistically significant trend
# Gray = "No significant trend" (CI crosses zero or p>=0.05)
# ============================================================

def make_forest_plot(df_apc, title, filename, sig_color, group_col="Species"):

    df_plot = df_apc.copy()

    # Extract APC and CI from string columns if needed
    if "APC % (95% CI)" in df_plot.columns:
        apc_vals = df_plot["APC % (95% CI)"].str.extract(
            r"([-\d.]+)\s*\(([-\d.]+)–([-\d.]+)\)"
        ).astype(float)
        df_plot["APC_val"] = apc_vals[0]
        df_plot["CI_L"]    = apc_vals[1]
        df_plot["CI_U"]    = apc_vals[2]
    else:
        df_plot["APC_val"] = df_plot["APC_%"]
        ci_split = df_plot["APC 95% CI"].str.split("–", expand=True)
        df_plot["CI_L"] = ci_split[0].astype(float)
        df_plot["CI_U"] = ci_split[1].astype(float)

    df_plot = df_plot.sort_values("APC_val")

    y_pos = np.arange(len(df_plot))
    colors = [
        sig_color if t != "No significant trend" else "#B0B0B0"
        for t in df_plot["Trend"]
    ]

    fig, ax = plt.subplots(figsize=(11, max(4, 0.55*len(df_plot))))

    ax.barh(
        y_pos, df_plot["APC_val"],
        xerr=[
            df_plot["APC_val"] - df_plot["CI_L"],
            df_plot["CI_U"] - df_plot["APC_val"]
        ],
        color=colors, alpha=0.85, capsize=4,
        error_kw={"linewidth": 1.4}
    )

    ax.axvline(0, color="black", linewidth=1.2, linestyle="--", alpha=0.6)
    ax.set_yticks(y_pos)
    ax.set_yticklabels(df_plot[group_col], fontsize=10)
    ax.set_xlabel("Annual Percentage Change (%)")
    ax.set_title(title, fontweight="bold", fontsize=12)
    ax.grid(axis="x", alpha=0.3)

    # Legend
    from matplotlib.patches import Patch
    legend_elems = [
        Patch(facecolor=sig_color, label="Statistically significant (p < 0.05)"),
        Patch(facecolor="#B0B0B0", label="No significant trend")
    ]
    ax.legend(handles=legend_elems, loc="lower right", fontsize=9, frameon=False)

    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()


# --- Carbapenem forest plot (species) ---
carb_species = species_df[species_df["Antibiotic"] == "Carbapenem"]
make_forest_plot(
    carb_species,
    "Carbapenem Resistance — APC by Species (95% CI)",
    "SO1_Block6_Fig2_Carbapenem_Forest.png",
    sig_color="#D62728"
)

# --- Cephalosporin forest plot (species) ---
ceph_species = species_df[species_df["Antibiotic"] == "3G Cephalosporin"]
make_forest_plot(
    ceph_species,
    "3G Cephalosporin Resistance — APC by Species (95% CI)",
    "SO1_Block6_Fig3_Cephalosporin_Forest.png",
    sig_color="#1F77B4"
)

print("✅ Figures 2–3 saved: species-level significance-aware forest plots")


# ============================================================
# SECTION 3: PATHOGEN-GROUP FOREST PLOT
# ============================================================

carb_group = stratified_df[stratified_df["Antibiotic Class"] == "Carbapenem"]
make_forest_plot(
    carb_group,
    "Carbapenem Resistance — APC by Pathogen Group (95% CI)",
    "SO1_Block6_Fig4_Group_Forest_Carbapenem.png",
    sig_color="#D62728",
    group_col="Pathogen Group"
)

ceph_group = stratified_df[stratified_df["Antibiotic Class"] == "Third-Gen Cephalosporin"]
make_forest_plot(
    ceph_group,
    "3G Cephalosporin Resistance — APC by Pathogen Group (95% CI)",
    "SO1_Block6_Fig5_Group_Forest_Cephalosporin.png",
    sig_color="#1F77B4",
    group_col="Pathogen Group"
)

print("✅ Figures 4–5 saved: pathogen-group forest plots")


# ============================================================
# SECTION 4: FINAL SIGNIFICANCE SUMMARY TABLE
# ============================================================
# One consolidated table — the single most reviewer-friendly
# output: what is real, and what is not.
# ============================================================

print("\n")
print("=" * 80)
print("TABLE 6: SIGNIFICANCE SUMMARY — WHAT IS A REAL TREND?")
print("=" * 80)

sig_summary = pd.concat([
    species_df[["Species", "Antibiotic", "APC % (95% CI)", "P-value", "Trend"]]
        .rename(columns={"Species": "Unit"}),
    stratified_df[["Pathogen Group", "Antibiotic Class", "APC % (95% CI)", "CA P-value", "Trend"]]
        .rename(columns={
            "Pathogen Group": "Unit",
            "Antibiotic Class": "Antibiotic",
            "CA P-value": "P-value"
        })
], ignore_index=True)

sig_summary["Statistically Real Trend?"] = sig_summary["Trend"].apply(
    lambda t: "❌ No" if t == "No significant trend" else "✅ Yes"
)

display(sig_summary)
sig_summary.to_csv("SO1_Block6_Significance_Summary.csv", index=False)


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 80)
print("BLOCK 6 COMPLETE — FILES SAVED")
print("=" * 80)
files = [
    "SO1_Block6_Fig1_PathogenGroup_Trends.png",
    "SO1_Block6_Fig2_Carbapenem_Forest.png",
    "SO1_Block6_Fig3_Cephalosporin_Forest.png",
    "SO1_Block6_Fig4_Group_Forest_Carbapenem.png",
    "SO1_Block6_Fig5_Group_Forest_Cephalosporin.png",
    "SO1_Block6_Significance_Summary.csv"
]
for f in files:
    print(f"  📄 {f}")
print("\n🏆 OBJECTIVE 1 IS NOW FULLY COMPLETE — statistics AND visuals")
print("   both correctly distinguish real trends from noise.")
print("=" * 80)
```

```python
# ============================================================
# SO1 — BLOCK 7: COUNTRY-LEVEL TREND ANALYSIS
# ============================================================
# Purpose: Identify which countries show the fastest-rising
# resistance, using the corrected 589,607-isolate dataset
# and the significance-aware APC function from Block 5.
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings("ignore")

required = ["df_gn", "calculate_annual_rates", "calculate_apc"]
missing_obj = [r for r in required if r not in globals()]
if missing_obj:
    raise ValueError(f"Missing from earlier blocks: {missing_obj}")

print("=" * 80)
print("SO1 — BLOCK 7: COUNTRY-LEVEL TREND ANALYSIS")
print("=" * 80)

# ------------------------------------------------------------
# 1. SELECT TOP 15 COUNTRIES BY ISOLATE VOLUME
# ------------------------------------------------------------
# Countries with very few isolates give unstable APC estimates,
# so we restrict to the top 15 for reliable trend estimation.

top_countries = df_gn["Country"].value_counts().head(15).index.tolist()
print(f"\nAnalyzing top {len(top_countries)} countries by isolate volume:")
print(top_countries)


# ------------------------------------------------------------
# 2. ANNUAL RATES + APC PER COUNTRY (ONE REUSABLE FUNCTION)
# ------------------------------------------------------------

def country_level_apc(df, countries, outcome_col):
    """
    Calculates annual resistance rate + APC + significance
    for each country, reusing the corrected calculate_apc()
    function from Block 5 (significance-aware).
    """
    rows = []
    annual_store = {}

    for country in countries:
        c_df = df[df["Country"] == country]
        annual = calculate_annual_rates(c_df, outcome_col)

        if len(annual) < 5:
            continue

        annual_store[country] = annual
        apc = calculate_apc(annual, "Resistance_Rate_%")

        if not apc:
            continue

        first = annual.iloc[0]
        last  = annual.iloc[-1]

        rows.append({
            "Country":              country,
            "N Isolates":           int(c_df[outcome_col].notna().sum()),
            "Rate 2004 % (95% CI)": f"{first['Resistance_Rate_%']:.1f} ({first['CI_Lower_%']:.1f}–{first['CI_Upper_%']:.1f})",
            "Rate 2024 % (95% CI)": f"{last['Resistance_Rate_%']:.1f} ({last['CI_Lower_%']:.1f}–{last['CI_Upper_%']:.1f})",
            "APC_%":                apc["APC_%"],
            "APC 95% CI":           f"{apc['CI_Lower_%']}–{apc['CI_Upper_%']}",
            "R_squared":            apc["R_squared"],
            "P_value":              "< 0.001" if apc["P_value"] < 0.001 else f"{apc['P_value']:.4f}",
            "Trend":                apc["Trend"]   # significance-aware
        })

    return pd.DataFrame(rows), annual_store


carb_country_df, carb_country_annual = country_level_apc(
    df_gn, top_countries, "Carbapenem_Resistant"
)
ceph_country_df, ceph_country_annual = country_level_apc(
    df_gn, top_countries, "Cephalosporin_Resistant"
)

print("\n")
print("=" * 80)
print("TABLE 7A: COUNTRY-LEVEL APC — CARBAPENEM RESISTANCE")
print("=" * 80)
display(carb_country_df.sort_values("APC_%", ascending=False))

print("\n")
print("=" * 80)
print("TABLE 7B: COUNTRY-LEVEL APC — 3G CEPHALOSPORIN RESISTANCE")
print("=" * 80)
display(ceph_country_df.sort_values("APC_%", ascending=False))


# ------------------------------------------------------------
# 3. GLOBAL HEATMAP — COUNTRY × YEAR
# ------------------------------------------------------------

def plot_country_heatmap(df, countries, outcome_col, title, filename):

    ctry_year = (
        df[df["Country"].isin(countries)]
        .groupby(["Country", "Year"])[outcome_col]
        .mean()
        .reset_index()
    )
    ctry_year["Rate_%"] = ctry_year[outcome_col] * 100

    heat = ctry_year.pivot(
        index="Country", columns="Year", values="Rate_%"
    ).loc[countries]

    plt.figure(figsize=(14, 8))
    sns.heatmap(
        heat, cmap="Reds", linewidths=0.3, linecolor="gray",
        cbar_kws={"label": "Resistance Rate (%)"}
    )
    plt.title(title, fontsize=13, fontweight="bold")
    plt.xlabel("Year")
    plt.ylabel("Country")
    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()

plot_country_heatmap(
    df_gn, top_countries, "Carbapenem_Resistant",
    "Carbapenem Resistance Heatmap — Top 15 Countries (2004–2024)",
    "SO1_Block7_Fig1_Carbapenem_Heatmap.png"
)

plot_country_heatmap(
    df_gn, top_countries, "Cephalosporin_Resistant",
    "3G Cephalosporin Resistance Heatmap — Top 15 Countries (2004–2024)",
    "SO1_Block7_Fig2_Cephalosporin_Heatmap.png"
)

print("✅ Figures saved: country heatmaps (carbapenem + cephalosporin)")

# ------------------------------------------------------------
# 4. SAVE TABLES
# ------------------------------------------------------------

carb_country_df.to_csv("SO1_Block7_Country_APC_Carbapenem.csv", index=False)
ceph_country_df.to_csv("SO1_Block7_Country_APC_Cephalosporin.csv", index=False)

print("\n")
print("=" * 80)
print("BLOCK 7 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# SO2 — BLOCK 1: DISTRIBUTION OF RESISTANCE ACROSS
# SPECIES, CLINICAL SOURCE, CARE SETTING, AND AGE GROUP
# ============================================================
#
# Objective: To analyze the distribution of resistance across
# countries, species, and clinical sources
#
# This block builds the categorical resistance-rate tables
# that will later justify feature selection for SO5 (ML model)
# ============================================================

import pandas as pd
import numpy as np
from scipy import stats
import warnings
warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# SAFETY CHECK
# ------------------------------------------------------------

required = ["df_gn"]
missing_obj = [r for r in required if r not in globals()]
if missing_obj:
    raise ValueError(f"Missing from earlier blocks: {missing_obj}")

for col in ["Carbapenem_Resistant", "Cephalosporin_Resistant"]:
    if col not in df_gn.columns:
        raise ValueError(f"{col} not found — rerun SO1 Blocks 1–2 first.")

print("=" * 80)
print("SO2 — BLOCK 1: DISTRIBUTION FOUNDATION")
print("=" * 80)
print(f"Dataset: {len(df_gn):,} isolates | "
      f"{df_gn['Species'].nunique()} species | "
      f"{df_gn['Country'].nunique()} countries")


# ============================================================
# SECTION 1: INSPECT RAW CATEGORICAL COLUMNS
# ============================================================
# We look at Source, Speciality, and Age Group BEFORE cleaning
# so we know exactly what we're working with.

print("\n" + "─" * 70)
print("RAW COLUMN INSPECTION")
print("─" * 70)

for col in ["Source", "Speciality", "Age Group", "Gender"]:
    print(f"\n{col} — unique values: {df_gn[col].nunique()}")
    print(f"  Missing: {df_gn[col].isna().sum():,} "
          f"({df_gn[col].isna().mean()*100:.1f}%)")
    print(f"  Top 10 values:")
    print(df_gn[col].value_counts().head(10).to_string())


# ============================================================
# SECTION 2: CLEAN AND GROUP SPECIMEN SOURCE
# ============================================================
# The ATLAS 'Source' column uses "Category:Subcategory" format
# e.g. "Blood:Blood", "Respiratory:Sputum", "Urinary:Urine"
# We extract the top-level category for meaningful grouping.

def extract_source_category(source_value):
    """
    Extracts the top-level specimen category from ATLAS
    'Source' values formatted as 'Category:Subcategory'.
    Falls back to the raw value if no colon is present.
    """
    if pd.isna(source_value):
        return np.nan
    if ":" in str(source_value):
        return str(source_value).split(":")[0].strip()
    return str(source_value).strip()

df_gn = df_gn.copy()
df_gn["Source_Category"] = df_gn["Source"].apply(extract_source_category)

print("\n" + "─" * 70)
print("SPECIMEN SOURCE — GROUPED CATEGORIES")
print("─" * 70)
print(df_gn["Source_Category"].value_counts())


# ============================================================
# SECTION 3: REUSABLE FUNCTION —
# RESISTANCE RATE BY CATEGORICAL VARIABLE
# ============================================================
# This mirrors the logic of calculate_annual_rates() from SO1,
# but groups by any categorical column instead of Year.

def resistance_by_category(df, category_col, outcome_col, min_n=100):
    """
    Calculates resistance rate + Wilson 95% CI for each
    category of a given column. Categories with fewer than
    min_n tested isolates are excluded for statistical stability.
    """
    results = []

    for category in df[category_col].dropna().unique():

        cat_df = df[df[category_col] == category]
        tested = cat_df[outcome_col].dropna()
        n = len(tested)

        if n < min_n:
            continue

        resistant = tested.sum()
        rate = resistant / n

        # Wilson score 95% CI (same method as SO1)
        z = 1.96
        denom = 1 + (z**2 / n)
        center = (rate + (z**2 / (2*n))) / denom
        margin = (z * np.sqrt((rate*(1-rate)/n) + (z**2/(4*n**2)))) / denom
        ci_l = max(0, center - margin) * 100
        ci_u = min(1, center + margin) * 100

        results.append({
            category_col:  category,
            "N_Tested":    n,
            "N_Resistant": int(resistant),
            "Resistance_Rate_%": round(rate * 100, 2),
            "CI_Lower_%":  round(ci_l, 2),
            "CI_Upper_%":  round(ci_u, 2)
        })

    return pd.DataFrame(results).sort_values(
        "Resistance_Rate_%", ascending=False
    ).reset_index(drop=True)


# ============================================================
# SECTION 4: RESISTANCE BY SPECIES (FORMALIZED)
# ============================================================

print("\n")
print("=" * 80)
print("TABLE 2.1A: CARBAPENEM RESISTANCE BY SPECIES")
print("=" * 80)

species_carb = resistance_by_category(
    df_gn, "Species", "Carbapenem_Resistant", min_n=200
)
display(species_carb)

print("\n")
print("=" * 80)
print("TABLE 2.1B: 3G CEPHALOSPORIN RESISTANCE BY SPECIES")
print("=" * 80)

species_ceph = resistance_by_category(
    df_gn, "Species", "Cephalosporin_Resistant", min_n=200
)
display(species_ceph)


# ============================================================
# SECTION 5: RESISTANCE BY SPECIMEN SOURCE
# ============================================================

print("\n")
print("=" * 80)
print("TABLE 2.2A: CARBAPENEM RESISTANCE BY SPECIMEN SOURCE")
print("=" * 80)

source_carb = resistance_by_category(
    df_gn, "Source_Category", "Carbapenem_Resistant", min_n=200
)
display(source_carb)

print("\n")
print("=" * 80)
print("TABLE 2.2B: 3G CEPHALOSPORIN RESISTANCE BY SPECIMEN SOURCE")
print("=" * 80)

source_ceph = resistance_by_category(
    df_gn, "Source_Category", "Cephalosporin_Resistant", min_n=200
)
display(source_ceph)


# ============================================================
# SECTION 6: RESISTANCE BY CLINICAL CARE SETTING (Speciality)
# ============================================================

print("\n")
print("=" * 80)
print("TABLE 2.3A: CARBAPENEM RESISTANCE BY CARE SETTING")
print("=" * 80)

setting_carb = resistance_by_category(
    df_gn, "Speciality", "Carbapenem_Resistant", min_n=200
)
display(setting_carb)

print("\n")
print("=" * 80)
print("TABLE 2.3B: 3G CEPHALOSPORIN RESISTANCE BY CARE SETTING")
print("=" * 80)

setting_ceph = resistance_by_category(
    df_gn, "Speciality", "Cephalosporin_Resistant", min_n=200
)
display(setting_ceph)


# ============================================================
# SECTION 7: RESISTANCE BY AGE GROUP
# ============================================================

print("\n")
print("=" * 80)
print("TABLE 2.4A: CARBAPENEM RESISTANCE BY AGE GROUP")
print("=" * 80)

age_carb = resistance_by_category(
    df_gn, "Age Group", "Carbapenem_Resistant", min_n=200
)
display(age_carb)

print("\n")
print("=" * 80)
print("TABLE 2.4B: 3G CEPHALOSPORIN RESISTANCE BY AGE GROUP")
print("=" * 80)

age_ceph = resistance_by_category(
    df_gn, "Age Group", "Cephalosporin_Resistant", min_n=200
)
display(age_ceph)


# ============================================================
# SECTION 8: SAVE ALL TABLES
# ============================================================

species_carb.to_csv("SO2_Block1_Species_Carbapenem.csv", index=False)
species_ceph.to_csv("SO2_Block1_Species_Cephalosporin.csv", index=False)
source_carb.to_csv("SO2_Block1_Source_Carbapenem.csv", index=False)
source_ceph.to_csv("SO2_Block1_Source_Cephalosporin.csv", index=False)
setting_carb.to_csv("SO2_Block1_Setting_Carbapenem.csv", index=False)
setting_ceph.to_csv("SO2_Block1_Setting_Cephalosporin.csv", index=False)
age_carb.to_csv("SO2_Block1_AgeGroup_Carbapenem.csv", index=False)
age_ceph.to_csv("SO2_Block1_AgeGroup_Cephalosporin.csv", index=False)

print("\n")
print("=" * 80)
print("BLOCK 1 COMPLETE — 8 TABLES SAVED")
print("=" * 80)
print("  📄 SO2_Block1_Species_Carbapenem.csv")
print("  📄 SO2_Block1_Species_Cephalosporin.csv")
print("  📄 SO2_Block1_Source_Carbapenem.csv")
print("  📄 SO2_Block1_Source_Cephalosporin.csv")
print("  📄 SO2_Block1_Setting_Carbapenem.csv")
print("  📄 SO2_Block1_Setting_Cephalosporin.csv")
print("  📄 SO2_Block1_AgeGroup_Carbapenem.csv")
print("  📄 SO2_Block1_AgeGroup_Cephalosporin.csv")
print("\nNext: Block 2 will add statistical significance testing")
print("(chi-square) and cross-tabulation heatmaps (species × source).")
```

```python
# ============================================================
# SAFETY FIX: DEFINE resistance_by_category LOCALLY
# (in case kernel reset since SO2 Block 1)
# ============================================================
def resistance_by_category(df, category_col, outcome_col, min_n=100):
    results = []
    for category in df[category_col].dropna().unique():
        cat_df = df[df[category_col] == category]
        tested = cat_df[outcome_col].dropna()
        n = len(tested)
        if n < min_n:
            continue
        resistant = tested.sum()
        rate = resistant / n
        z = 1.96
        denom = 1 + (z**2 / n)
        center = (rate + (z**2 / (2*n))) / denom
        margin = (z * np.sqrt((rate*(1-rate)/n) + (z**2/(4*n**2)))) / denom
        results.append({
            category_col: category, "N_Tested": n, "N_Resistant": int(resistant),
            "Resistance_Rate_%": round(rate * 100, 2),
            "CI_Lower_%": round(max(0, center-margin)*100, 2),
            "CI_Upper_%": round(min(1, center+margin)*100, 2)
        })
    return pd.DataFrame(results).sort_values("Resistance_Rate_%", ascending=False).reset_index(drop=True)

setting_carb = resistance_by_category(df_gn, "Speciality", "Carbapenem_Resistant", min_n=200)
setting_ceph = resistance_by_category(df_gn, "Speciality", "Cephalosporin_Resistant", min_n=200)
print("✅ setting_carb/setting_ceph created")
```

```python
# ============================================================
# SO2 — BLOCK 2 (CORRECTED): CHI-SQUARE SIGNIFICANCE TESTING +
# CROSS-TABULATION HEATMAPS (SPECIES × SOURCE, SPECIES × SETTING)
# ============================================================
#
# Purpose: Statistically confirm that resistance differs
# significantly across species, specimen source, care setting,
# and age group — establishing evidence for feature relevance
# ahead of SO5 (ML prediction).
#
# FIX APPLIED: Section 5 no longer depends on a saved CSV file
# (which caused a FileNotFoundError on kernel reset). It now
# reuses in-memory tables from Block 1, with a safe fallback
# that recomputes them if needed.
# ============================================================

import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

required = ["df_gn"]
missing_obj = [r for r in required if r not in globals()]
if missing_obj:
    raise ValueError(f"Missing from earlier blocks: {missing_obj}")

print("=" * 80)
print("SO2 — BLOCK 2 (CORRECTED): STATISTICAL TESTING + CROSS-TABULATION")
print("=" * 80)

# ============================================================
# SECTION 1: CHI-SQUARE TEST — REUSABLE FUNCTION
# ============================================================
def chi_square_association(df, category_col, outcome_col, min_n=200):
    """
    Builds a contingency table (category × Resistant/Not-Resistant)
    and runs a chi-square test of independence.
    Categories with fewer than min_n tested isolates are excluded.
    Missing values in either column are dropped before testing.
    """
    work = df[[category_col, outcome_col]].dropna().copy()

    counts = work[category_col].value_counts()
    valid_categories = counts[counts >= min_n].index
    work = work[work[category_col].isin(valid_categories)]

    if work[category_col].nunique() < 2:
        return None

    contingency = pd.crosstab(work[category_col], work[outcome_col])
    chi2, p, dof, expected = stats.chi2_contingency(contingency)

    n = contingency.values.sum()
    min_dim = min(contingency.shape) - 1
    cramers_v = np.sqrt(chi2 / (n * min_dim)) if min_dim > 0 else np.nan

    return {
        "Variable": category_col,
        "N Categories": work[category_col].nunique(),
        "N Isolates": n,
        "Chi2_statistic": round(chi2, 2),
        "df": dof,
        "P_value": p,
        "Cramers_V": round(cramers_v, 4),
        "Effect_Size": (
            "Negligible" if cramers_v < 0.1 else
            "Small" if cramers_v < 0.3 else
            "Medium" if cramers_v < 0.5 else
            "Large"
        ),
        "Significant": p < 0.05
    }

def format_p(p):
    return "< 0.001" if p < 0.001 else f"{p:.4f}"

# ============================================================
# SECTION 2: RUN CHI-SQUARE TESTS — ALL VARIABLES
# ============================================================
test_configs = [
    ("Species", "Carbapenem_Resistant"),
    ("Species", "Cephalosporin_Resistant"),
    ("Source_Category", "Carbapenem_Resistant"),
    ("Source_Category", "Cephalosporin_Resistant"),
    ("Speciality", "Carbapenem_Resistant"),
    ("Speciality", "Cephalosporin_Resistant"),
    ("Age Group", "Carbapenem_Resistant"),
    ("Age Group", "Cephalosporin_Resistant"),
]

chi_results = []
print("\n" + "─" * 70)
print("CHI-SQUARE TESTS OF INDEPENDENCE")
print("─" * 70)

for category_col, outcome_col in test_configs:
    result = chi_square_association(df_gn, category_col, outcome_col)
    if result is None:
        continue

    ab_label = "Carbapenem" if "Carbapenem" in outcome_col else "3G Cephalosporin"
    print(f"\n{category_col} vs {ab_label} Resistance:")
    print(f"  χ² = {result['Chi2_statistic']}, df = {result['df']}, p {format_p(result['P_value'])}")
    print(f"  Cramér's V = {result['Cramers_V']} ({result['Effect_Size']} effect)")
    print(f"  Significant: {'Yes' if result['Significant'] else 'No'}")

    chi_results.append({
        "Variable": category_col, "Antibiotic": ab_label,
        "N Categories": result["N Categories"], "N Isolates": result["N Isolates"],
        "Chi2": result["Chi2_statistic"], "df": result["df"],
        "P-value": format_p(result["P_value"]), "Cramers_V": result["Cramers_V"],
        "Effect Size": result["Effect_Size"],
        "Significant": "Yes" if result["Significant"] else "No"
    })

chi_summary_df = pd.DataFrame(chi_results)
print("\n" + "=" * 80)
print("TABLE 2.5: CHI-SQUARE SUMMARY — ALL VARIABLES")
print("=" * 80)
display(chi_summary_df)
chi_summary_df.to_csv("SO2_Block2_ChiSquare_Summary.csv", index=False)

# ============================================================
# SECTION 3: HEATMAP — WHO PRIORITY SPECIES × CARE SETTING
# ============================================================
who_priority_species = [
    "Escherichia coli", "Klebsiella pneumoniae", "Pseudomonas aeruginosa",
    "Acinetobacter baumannii", "Enterobacter cloacae", "Proteus mirabilis",
    "Serratia marcescens", "Klebsiella oxytoca", "Citrobacter freundii",
    "Morganella morganii"
]

def build_heatmap_matrix(df, row_col, row_values, col_col, outcome_col, min_n=30):
    subset = df[df[row_col].isin(row_values)]
    matrix = subset.groupby([row_col, col_col])[outcome_col].agg(["mean", "count"]).reset_index()
    matrix.loc[matrix["count"] < min_n, "mean"] = np.nan
    matrix["Rate_%"] = matrix["mean"] * 100
    pivot = matrix.pivot(index=row_col, columns=col_col, values="Rate_%")
    pivot = pivot.reindex([s for s in row_values if s in pivot.index])
    return pivot

fig, axes = plt.subplots(1, 2, figsize=(20, 8))

carb_matrix = build_heatmap_matrix(df_gn, "Species", who_priority_species, "Speciality", "Carbapenem_Resistant")
sns.heatmap(carb_matrix, annot=True, fmt=".1f", cmap="Reds", linewidths=0.4, linecolor="gray",
            cbar_kws={"label": "Resistance Rate (%)"}, ax=axes[0])
axes[0].set_title("A. Carbapenem Resistance\nWHO Priority Species × Care Setting", fontweight="bold", fontsize=12)
axes[0].set_xlabel("Care Setting")
axes[0].set_ylabel("Species")
plt.setp(axes[0].get_xticklabels(), rotation=45, ha="right")

ceph_matrix = build_heatmap_matrix(df_gn, "Species", who_priority_species, "Speciality", "Cephalosporin_Resistant")
sns.heatmap(ceph_matrix, annot=True, fmt=".1f", cmap="Blues", linewidths=0.4, linecolor="gray",
            cbar_kws={"label": "Resistance Rate (%)"}, ax=axes[1])
axes[1].set_title("B. 3G Cephalosporin Resistance\nWHO Priority Species × Care Setting", fontweight="bold", fontsize=12)
axes[1].set_xlabel("Care Setting")
axes[1].set_ylabel("Species")
plt.setp(axes[1].get_xticklabels(), rotation=45, ha="right")

plt.suptitle("Cross-Tabulated Resistance Patterns — Species × Clinical Setting\nATLAS Global Surveillance 2004–2024",
              fontsize=14, fontweight="bold", y=1.03)
plt.tight_layout()
plt.savefig("SO2_Block2_Fig1_Species_x_Setting_Heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
print("✅ Figure 1 saved: species × care setting heatmap")

# ============================================================
# SECTION 4: HEATMAP — WHO PRIORITY SPECIES × TOP 10 SOURCES
# ============================================================
top_10_sources = df_gn["Source_Category"].value_counts().head(10).index.tolist()
print(f"\nTop 10 specimen sources used for heatmap: {top_10_sources}")

fig2, axes2 = plt.subplots(1, 2, figsize=(20, 8))

carb_source_matrix = build_heatmap_matrix(
    df_gn[df_gn["Source_Category"].isin(top_10_sources)],
    "Species", who_priority_species, "Source_Category", "Carbapenem_Resistant"
)
sns.heatmap(carb_source_matrix, annot=True, fmt=".1f", cmap="Reds", linewidths=0.4, linecolor="gray",
            cbar_kws={"label": "Resistance Rate (%)"}, ax=axes2[0])
axes2[0].set_title("A. Carbapenem Resistance\nWHO Priority Species × Top 10 Specimen Sources", fontweight="bold", fontsize=12)
axes2[0].set_xlabel("Specimen Source")
axes2[0].set_ylabel("Species")
plt.setp(axes2[0].get_xticklabels(), rotation=45, ha="right")

ceph_source_matrix = build_heatmap_matrix(
    df_gn[df_gn["Source_Category"].isin(top_10_sources)],
    "Species", who_priority_species, "Source_Category", "Cephalosporin_Resistant"
)
sns.heatmap(ceph_source_matrix, annot=True, fmt=".1f", cmap="Blues", linewidths=0.4, linecolor="gray",
            cbar_kws={"label": "Resistance Rate (%)"}, ax=axes2[1])
axes2[1].set_title("B. 3G Cephalosporin Resistance\nWHO Priority Species × Top 10 Specimen Sources", fontweight="bold", fontsize=12)
axes2[1].set_xlabel("Specimen Source")
axes2[1].set_ylabel("Species")
plt.setp(axes2[1].get_xticklabels(), rotation=45, ha="right")

plt.suptitle("Cross-Tabulated Resistance Patterns — Species × Specimen Source\nATLAS Global Surveillance 2004–2024",
              fontsize=14, fontweight="bold", y=1.03)
plt.tight_layout()
plt.savefig("SO2_Block2_Fig2_Species_x_Source_Heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
print("✅ Figure 2 saved: species × specimen source heatmap")

# ============================================================
# SECTION 5 (CORRECTED): BAR CHART — CARE SETTING COMPARISON
# Fix: no longer depends on a saved CSV file. Uses in-memory
# tables from Block 1, with a safe fallback that recomputes
# them if the kernel was reset since Block 1 last ran.
# ============================================================
if "setting_carb" not in globals() or "setting_ceph" not in globals():
    print("\n⚠️  setting_carb/setting_ceph not found in memory — recomputing...")
    setting_carb = resistance_by_category(df_gn, "Speciality", "Carbapenem_Resistant", min_n=200)
    setting_ceph = resistance_by_category(df_gn, "Speciality", "Cephalosporin_Resistant", min_n=200)
else:
    print("\n✅ Using setting_carb/setting_ceph already in memory from Block 1")

fig3, axes3 = plt.subplots(1, 2, figsize=(18, 7))

for ax, data, color, title in zip(
    axes3, [setting_carb, setting_ceph], ["#D62728", "#1F77B4"],
    ["A. Carbapenem Resistance by Care Setting", "B. 3G Cephalosporin Resistance by Care Setting"]
):
    data_sorted = data.sort_values("Resistance_Rate_%", ascending=True)
    ax.barh(
        data_sorted["Speciality"], data_sorted["Resistance_Rate_%"],
        xerr=[data_sorted["Resistance_Rate_%"] - data_sorted["CI_Lower_%"],
              data_sorted["CI_Upper_%"] - data_sorted["Resistance_Rate_%"]],
        color=color, alpha=0.8, capsize=3
    )
    ax.set_xlabel("Resistance Rate (%)")
    ax.set_title(title, fontweight="bold", fontsize=12)
    ax.grid(axis="x", alpha=0.3)

plt.suptitle("Resistance Rate by Clinical Care Setting (95% CI)\nWHO Priority Gram-Negative Pathogens — ATLAS 2004–2024",
              fontsize=14, fontweight="bold", y=1.03)
plt.tight_layout()
plt.savefig("SO2_Block2_Fig3_CareSetting_Comparison.png", dpi=300, bbox_inches="tight")
plt.show()
print("✅ Figure 3 saved: care setting comparison bar chart")

# ============================================================
# FINAL SUMMARY
# ============================================================
print("\n" + "=" * 80)
print("BLOCK 2 (CORRECTED) COMPLETE")
print("=" * 80)
print("  📄 SO2_Block2_ChiSquare_Summary.csv")
print("  🖼️  SO2_Block2_Fig1_Species_x_Setting_Heatmap.png")
print("  🖼️  SO2_Block2_Fig2_Species_x_Source_Heatmap.png")
print("  🖼️  SO2_Block2_Fig3_CareSetting_Comparison.png")
print("=" * 80)
```

# Phenotypic susceptibility patterns

```python
# ============================================================
# SO3 — BLOCK 1: PHENOTYPIC SUSCEPTIBILITY PATTERNS (S/I/R)
# ============================================================
#
# Objective: Evaluate phenotypic susceptibility patterns
# (Susceptible / Intermediate / Resistant) for key antibiotics
#
# This reuses df_gn and the same 9 individual drug columns
# already validated throughout SO1/SO2/SO5 — no new data,
# just a different, more granular view of the same columns.
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# SAFETY CHECK
# ------------------------------------------------------------
required = ["df_gn"]
missing_obj = [r for r in required if r not in globals()]
if missing_obj:
    raise ValueError(f"Missing from earlier blocks: {missing_obj}")

print("=" * 80)
print("SO3 — BLOCK 1: PHENOTYPIC SUSCEPTIBILITY PATTERNS (S/I/R)")
print("=" * 80)
print(f"Dataset: {len(df_gn):,} isolates | {df_gn['Species'].nunique()} species")

# ------------------------------------------------------------
# 1. DEFINE THE 9 KEY DRUGS (same ones used throughout SO1/SO2/SO5)
# ------------------------------------------------------------
key_drugs = {
    "Meropenem_I":     "Carbapenem",
    "Imipenem_I":      "Carbapenem",
    "Doripenem_I":     "Carbapenem",
    "Ertapenem_I":     "Carbapenem",
    "Ceftriaxone_I":   "3G Cephalosporin",
    "Ceftazidime_I":   "3G Cephalosporin",
    "Cefpodoxime_I":   "3G Cephalosporin",
    "Cefixime_I":      "3G Cephalosporin",
    "Ceftibuten_I":    "3G Cephalosporin",
}

key_drugs = {d: cls for d, cls in key_drugs.items() if d in df_gn.columns}

# ------------------------------------------------------------
# 2. FULL S/I/R BREAKDOWN PER DRUG
# ------------------------------------------------------------
def sir_breakdown(df, drug_col):
    """
    Calculates the full Susceptible/Intermediate/Resistant
    breakdown for one drug column. Missing values (isolates
    not tested with this specific drug) are excluded via
    value_counts()'s default behavior, which ignores NaN.
    """
    counts = df[drug_col].value_counts()
    n_tested = counts.get("Susceptible", 0) + counts.get("Intermediate", 0) + counts.get("Resistant", 0)

    if n_tested == 0:
        return None

    return {
        "Drug": drug_col.replace("_I", ""),
        "N_Tested": n_tested,
        "Susceptible_%": round(counts.get("Susceptible", 0) / n_tested * 100, 2),
        "Intermediate_%": round(counts.get("Intermediate", 0) / n_tested * 100, 2),
        "Resistant_%": round(counts.get("Resistant", 0) / n_tested * 100, 2),
        "N_Susceptible": int(counts.get("Susceptible", 0)),
        "N_Intermediate": int(counts.get("Intermediate", 0)),
        "N_Resistant": int(counts.get("Resistant", 0)),
    }

print("\n" + "─" * 70)
print("BUILDING ANTIBIOGRAM — S/I/R BREAKDOWN PER DRUG")
print("─" * 70)

antibiogram_rows = []
for drug_col, drug_class in key_drugs.items():
    result = sir_breakdown(df_gn, drug_col)
    if result:
        result["Class"] = drug_class
        antibiogram_rows.append(result)

antibiogram_df = pd.DataFrame(antibiogram_rows)
antibiogram_df = antibiogram_df[
    ["Drug", "Class", "N_Tested", "Susceptible_%", "Intermediate_%", "Resistant_%",
     "N_Susceptible", "N_Intermediate", "N_Resistant"]
].sort_values("Resistant_%", ascending=False).reset_index(drop=True)

print("\n" + "=" * 80)
print("TABLE 3.1: ANTIBIOGRAM — SUSCEPTIBILITY PATTERNS BY DRUG")
print("=" * 80)
display(antibiogram_df)
antibiogram_df.to_csv("SO3_Block1_Antibiogram_Summary.csv", index=False)

# ------------------------------------------------------------
# 3. ANTIBIOGRAM VISUALIZATION — STACKED BAR CHART
# ------------------------------------------------------------
# Standard clinical microbiology antibiogram format:
# each drug as a horizontal bar, split into S/I/R proportions

plot_df = antibiogram_df.sort_values("Resistant_%", ascending=True)

fig, ax = plt.subplots(figsize=(11, 6))

ax.barh(plot_df["Drug"], plot_df["Susceptible_%"], color="#2CA02C", label="Susceptible")
ax.barh(plot_df["Drug"], plot_df["Intermediate_%"], left=plot_df["Susceptible_%"],
        color="#FF7F0E", label="Intermediate")
ax.barh(plot_df["Drug"], plot_df["Resistant_%"],
        left=plot_df["Susceptible_%"] + plot_df["Intermediate_%"],
        color="#D62728", label="Resistant")

ax.set_xlabel("Percentage of Isolates Tested (%)")
ax.set_title("Antibiogram — Phenotypic Susceptibility Patterns\nWHO Priority Gram-Negative Pathogens, ATLAS 2004–2024",
              fontweight="bold", fontsize=13)
ax.legend(loc="lower right", frameon=False)
ax.set_xlim(0, 100)
ax.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.savefig("SO3_Block1_Fig1_Antibiogram.png", dpi=300, bbox_inches="tight")
plt.show()
print("\n✅ Figure saved: antibiogram (S/I/R stacked bar chart)")

# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------
print("\n" + "=" * 80)
print("BLOCK 1 COMPLETE")
print("=" * 80)
print("  📄 SO3_Block1_Antibiogram_Summary.csv")
print("  🖼️  SO3_Block1_Fig1_Antibiogram.png")
print("\nNext: Block 2 will break this down by WHO-priority species,")
print("showing which drugs remain most reliably effective per pathogen.")
```

```python
# ============================================================
# SO3 — BLOCK 2 (FULLY CONSOLIDATED): SPECIES-LEVEL
# SUSCEPTIBILITY PATTERNS, WITH RELIABILITY CHECK BUILT IN
# ============================================================
#
# This single block replaces all earlier separate pieces.
# It produces:
#   1. Species × Drug resistance matrix (Table 3.2)
#   2. Heatmap visualization
#   3. Most/Least effective drug per species — WITH sample
#      size and a reliability check against Meropenem
#      (the most completely-tested drug), so you know which
#      results to trust and which to treat cautiously.
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

required = ["df_gn"]
missing_obj = [r for r in required if r not in globals()]
if missing_obj:
    raise ValueError(f"Missing from earlier blocks: {missing_obj}")

print("=" * 80)
print("SO3 — BLOCK 2: SPECIES-LEVEL SUSCEPTIBILITY PATTERNS (CONSOLIDATED)")
print("=" * 80)

# ------------------------------------------------------------
# 1. SETUP — drugs and species
# ------------------------------------------------------------
key_drugs = {
    "Meropenem_I": "Carbapenem", "Imipenem_I": "Carbapenem",
    "Doripenem_I": "Carbapenem", "Ertapenem_I": "Carbapenem",
    "Ceftriaxone_I": "3G Cephalosporin", "Ceftazidime_I": "3G Cephalosporin",
    "Cefpodoxime_I": "3G Cephalosporin", "Cefixime_I": "3G Cephalosporin",
    "Ceftibuten_I": "3G Cephalosporin",
}
key_drugs = {d: cls for d, cls in key_drugs.items() if d in df_gn.columns}

who_priority_species = [
    "Escherichia coli", "Klebsiella pneumoniae", "Pseudomonas aeruginosa",
    "Acinetobacter baumannii", "Enterobacter cloacae", "Proteus mirabilis",
    "Serratia marcescens", "Klebsiella oxytoca", "Citrobacter freundii",
    "Morganella morganii"
]

# ------------------------------------------------------------
# 2. BUILD THE SPECIES × DRUG RESISTANCE MATRIX
# ------------------------------------------------------------
def species_drug_resistance(df, species_list, drug_col, min_n=30):
    rows = []
    for species in species_list:
        sp_df = df[df["Species"] == species]
        tested = sp_df[drug_col].dropna()
        n = len(tested)
        if n < min_n:
            rows.append({"Species": species, "N_Tested": n, "Resistant_%": np.nan})
            continue
        resistant_pct = (tested == "Resistant").sum() / n * 100
        rows.append({"Species": species, "N_Tested": n, "Resistant_%": round(resistant_pct, 2)})
    return pd.DataFrame(rows)

print("\nStep 1 — Building species × drug resistance matrix...")
matrix_rows = []
for drug_col, drug_class in key_drugs.items():
    result = species_drug_resistance(df_gn, who_priority_species, drug_col)
    result["Drug"] = drug_col.replace("_I", "")
    matrix_rows.append(result)
species_drug_df = pd.concat(matrix_rows, ignore_index=True)

resistance_pivot = species_drug_df.pivot(index="Species", columns="Drug", values="Resistant_%")
resistance_pivot = resistance_pivot.reindex(who_priority_species)
drug_order = ["Ertapenem", "Doripenem", "Meropenem", "Imipenem",
              "Ceftibuten", "Ceftazidime", "Ceftriaxone", "Cefpodoxime", "Cefixime"]
drug_order = [d for d in drug_order if d in resistance_pivot.columns]
resistance_pivot = resistance_pivot[drug_order]

print("\n" + "=" * 80)
print("TABLE 3.2 — RESISTANCE % BY SPECIES × DRUG")
print("(Each cell = % of isolates of that species resistant to that drug)")
print("=" * 80)
display(resistance_pivot)
resistance_pivot.to_csv("SO3_Block2_Species_Drug_Resistance_Matrix.csv")

# ------------------------------------------------------------
# 3. HEATMAP
# ------------------------------------------------------------
fig, ax = plt.subplots(figsize=(13, 8))
sns.heatmap(
    resistance_pivot, annot=True, fmt=".1f", cmap="Reds",
    linewidths=0.4, linecolor="gray", vmin=0, vmax=100,
    cbar_kws={"label": "Resistance Rate (%)"}, ax=ax
)
ax.set_title(
    "Species-Level Susceptibility Patterns — WHO Priority Pathogens\nATLAS Global Surveillance 2004–2024",
    fontweight="bold", fontsize=13
)
ax.set_xlabel("Antibiotic")
ax.set_ylabel("Species")
plt.setp(ax.get_xticklabels(), rotation=45, ha="right")
plt.tight_layout()
plt.savefig("SO3_Block2_Fig1_Species_Drug_Heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
print("✅ Figure saved: species × drug resistance heatmap")

# ------------------------------------------------------------
# 4. BEST/WORST DRUG PER SPECIES — WITH RELIABILITY CHECK
# ------------------------------------------------------------
print("\nStep 2 — Identifying the most/least effective drug per species,")
print("and checking each 'winner' against Meropenem (the most heavily-")
print("tested drug overall) to catch any small-sample-size illusions.\n")

best_drug_rows = []
for species in who_priority_species:
    if species not in resistance_pivot.index:
        continue
    row = resistance_pivot.loc[species].dropna()
    if row.empty:
        continue

    best_drug = row.idxmin()
    best_rate = row.min()
    worst_drug = row.idxmax()
    worst_rate = row.max()

    n_best = species_drug_df[(species_drug_df["Species"] == species) & (species_drug_df["Drug"] == best_drug)]["N_Tested"].values[0]

    mero_row = species_drug_df[(species_drug_df["Species"] == species) & (species_drug_df["Drug"] == "Meropenem")]
    mero_rate = mero_row["Resistant_%"].values[0] if not mero_row.empty else np.nan

    # Only flag as "check carefully" if BOTH drugs are carbapenems
    # (fair, same-class comparison) AND the gap is large
    same_class_gap = abs(best_rate - mero_rate) if best_drug in ["Doripenem", "Ertapenem", "Imipenem"] else 0
    reliability = "⚠️ Check — diverges from Meropenem" if same_class_gap >= 5 else "✅ Consistent"

    best_drug_rows.append({
        "Species": species,
        "Most Effective Drug": best_drug,
        "Resistance %": best_rate,
        "N Tested": n_best,
        "Meropenem Resistance % (reference)": mero_rate,
        "Reliability": reliability,
        "Least Effective Drug": worst_drug,
        "Resistance % (worst)": worst_rate,
    })

best_drug_df = pd.DataFrame(best_drug_rows)
print("=" * 80)
print("TABLE 3.3 — MOST/LEAST EFFECTIVE DRUG PER SPECIES (RELIABILITY-CHECKED)")
print("=" * 80)
display(best_drug_df)
best_drug_df.to_csv("SO3_Block2_Best_Worst_Drug_Per_Species.csv", index=False)

# ------------------------------------------------------------
# PLAIN-LANGUAGE SUMMARY
# ------------------------------------------------------------
print("\n" + "=" * 80)
print("WHAT THIS MEANS — PLAIN SUMMARY")
print("=" * 80)
print("• Table 3.2 shows raw resistance % for every species-drug pair.")
print("• The heatmap is the same information, visualized (darker red = more resistant).")
print("• Table 3.3 answers: 'which drug still works best/worst against each pathogen?'")
print("  — with a built-in check flagging any result that might just be due to a")
print("    small number of tested isolates, rather than genuine drug effectiveness.")
print("=" * 80)
```

```python
# ============================================================
# SO3 — BLOCK 3: INTERMEDIATE SUSCEPTIBILITY ANALYSIS
# ============================================================
#
# Objective: Complete the S/I/R picture by examining which
# drugs/species show meaningful "Intermediate" rates — a
# clinically important category often overlooked when only
# "Resistant %" is reported.
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

required = ["df_gn"]
missing_obj = [r for r in required if r not in globals()]
if missing_obj:
    raise ValueError(f"Missing from earlier blocks: {missing_obj}")

print("=" * 80)
print("SO3 — BLOCK 3: INTERMEDIATE SUSCEPTIBILITY ANALYSIS")
print("=" * 80)
print("\nWhy this matters: 'Intermediate' isolates are not simply")
print("'partially resistant' — clinically, they may still respond")
print("to treatment at higher doses. Reporting only 'Resistant %'")
print("can understate how many isolates need special dosing")
print("consideration rather than being untreatable.\n")

key_drugs = {
    "Meropenem_I": "Carbapenem", "Imipenem_I": "Carbapenem",
    "Doripenem_I": "Carbapenem", "Ertapenem_I": "Carbapenem",
    "Ceftriaxone_I": "3G Cephalosporin", "Ceftazidime_I": "3G Cephalosporin",
    "Cefpodoxime_I": "3G Cephalosporin", "Cefixime_I": "3G Cephalosporin",
    "Ceftibuten_I": "3G Cephalosporin",
}
key_drugs = {d: cls for d, cls in key_drugs.items() if d in df_gn.columns}

# ------------------------------------------------------------
# 1. WHICH DRUGS HAVE THE HIGHEST INTERMEDIATE RATES?
# ------------------------------------------------------------
intermediate_rows = []
for drug_col, drug_class in key_drugs.items():
    counts = df_gn[drug_col].value_counts()
    n_tested = counts.get("Susceptible", 0) + counts.get("Intermediate", 0) + counts.get("Resistant", 0)
    if n_tested == 0:
        continue
    intermediate_rows.append({
        "Drug": drug_col.replace("_I", ""),
        "Class": drug_class,
        "N_Tested": n_tested,
        "Intermediate_%": round(counts.get("Intermediate", 0) / n_tested * 100, 2),
        "N_Intermediate": int(counts.get("Intermediate", 0)),
    })

intermediate_df = pd.DataFrame(intermediate_rows).sort_values("Intermediate_%", ascending=False).reset_index(drop=True)

print("=" * 80)
print("TABLE 3.4: INTERMEDIATE SUSCEPTIBILITY RATE BY DRUG")
print("(Ranked highest to lowest — these are candidates for")
print(" dose-optimization discussion rather than \"untreatable\")")
print("=" * 80)
display(intermediate_df)
intermediate_df.to_csv("SO3_Block3_Intermediate_By_Drug.csv", index=False)

# ------------------------------------------------------------
# 2. VISUALIZATION — INTERMEDIATE VS RESISTANT, SIDE BY SIDE
# ------------------------------------------------------------
full_df = pd.read_csv("SO3_Block1_Antibiogram_Summary.csv") if "antibiogram_df" not in globals() else antibiogram_df
plot_df = full_df.merge(intermediate_df[["Drug", "Intermediate_%"]], on="Drug", suffixes=("", "_check"))
plot_df = plot_df.sort_values("Resistant_%", ascending=True)

fig, ax = plt.subplots(figsize=(11, 6))
x = np.arange(len(plot_df))
width = 0.35
ax.barh(x - width/2, plot_df["Intermediate_%"], height=width, color="#FF7F0E", label="Intermediate")
ax.barh(x + width/2, plot_df["Resistant_%"], height=width, color="#D62728", label="Resistant")
ax.set_yticks(x)
ax.set_yticklabels(plot_df["Drug"])
ax.set_xlabel("Percentage of Isolates Tested (%)")
ax.set_title("Intermediate vs. Resistant Rates by Drug\nWHO Priority Gram-Negative Pathogens, ATLAS 2004–2024",
              fontweight="bold", fontsize=13)
ax.legend(loc="lower right", frameon=False)
ax.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.savefig("SO3_Block3_Fig1_Intermediate_vs_Resistant.png", dpi=300, bbox_inches="tight")
plt.show()
print("\n✅ Figure saved: Intermediate vs Resistant comparison")

# ------------------------------------------------------------
# PLAIN SUMMARY
# ------------------------------------------------------------
top_intermediate = intermediate_df.iloc[0]
print("\n" + "=" * 80)
print("KEY TAKEAWAY")
print("=" * 80)
print(f"'{top_intermediate['Drug']}' has the highest Intermediate rate "
      f"({top_intermediate['Intermediate_%']}%) — isolates in this category")
print("are not fully resistant, and may still be treatable with adjusted")
print("dosing strategies. This nuance is lost if only 'Resistant %' is reported.")
print("=" * 80)

# ------------------------------------------------------------
# FINAL SO3 STATUS CHECK
# ------------------------------------------------------------
print("\n" + "=" * 80)
print("🏆 OBJECTIVE 3 (PHENOTYPIC SUSCEPTIBILITY PATTERNS) — STATUS")
print("=" * 80)
print("  ✅ Overall antibiogram (S/I/R per drug) — Block 1")
print("  ✅ Species-level susceptibility patterns — Block 2")
print("  ✅ Intermediate category analysis — Block 3")
print("\nSO3 is now complete, matching the same rigor as SO1/SO2/SO5.")
```
